```python
import context_cite
from transformers import AutoTokenizer

from context_cite import ContextCiter
```

```python
import pandas as pd

# Load the CSV file
df = pd.read_csv("merged_llm_4k_questions.csv")

# Display the first few rows after removing duplicates
# print(df_op4.columns)
# print(df_op4.head())
print(len(df))
```

```python
model_id = "deepseek-ai/DeepSeek-R1-Distill-Qwen-14B"
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM    
model = AutoModelForCausalLM.from_pretrained(
        model_id,
        torch_dtype=torch.float16,
        device_map="auto"  # ✅ valid here
    )
tokenizer = AutoTokenizer.from_pretrained(model_id)
```

```python
import os
import re
import pandas as pd
from transformers import AutoTokenizer, AutoModelForCausalLM

# Create output directory
output_dir = str(paths.CONTEXTCITE_14B_STAGING)
os.makedirs(output_dir, exist_ok=True)

# Loop through the first 298 rows
for idx, row in df.head(4054).iterrows():
    context_text = row["context"]
    question = row["question"]  # ✅ define this first

    # Append instruction to query
    query_full = (
        f"{question}\n"
        "Read the question and state your answer. "
        "State your answer, starting with 'Answer:', ending with two line breaks.\n\n"
    )
    
    try:
        # Generate attribution and response
        cc = ContextCiter(
            model,
            tokenizer,
            context=context_text,
            query=query_full,
            generate_kwargs={"max_new_tokens": 2048, "do_sample": False, "pad_token_id": tokenizer.eos_token_id}
        )

        # Extract model response
        raw_response = cc.response.strip()
        match = re.search(r"Answer:\s*([A-J])", raw_response)
        extracted_answer = match.group(1).strip() if match else None

        # Get attribution scores
        result = cc.get_attributions(as_dataframe=True)
        if isinstance(result, pd.io.formats.style.Styler):
            result = result.data

        if isinstance(result, pd.DataFrame):
            qa_id = f"Merge Q{idx + 1}"
            result["QA_ID"] = qa_id
            result["Extracted_Answer"] = extracted_answer

            # Save CSV
            filename = f"{qa_id}.csv".replace(" ", "_")
            result.to_csv(os.path.join(output_dir, filename), index=False)

            print(f"✅ Saved attribution for {qa_id}")
        else:
            print(f"⚠️ Unexpected result type at row {idx}: {type(result)}")

    except Exception as e:
        print(f"❌ Error on row {idx}: {e}")
```

```python
import os
import re
import pandas as pd
from transformers import AutoTokenizer, AutoModelForCausalLM

# Create output directory
output_dir = str(paths.CONTEXTCITE_14B_STAGING)
os.makedirs(output_dir, exist_ok=True)

# Get already processed QA_IDs from filenames
processed_ids = {
    f.replace(".csv", "").replace("_", " ")  # convert filename to QA_ID format
    for f in os.listdir(output_dir)
    if f.endswith(".csv")
}

# Loop through DataFrame rows
for idx, row in df.head(4053).iterrows():
    qa_id = f"Merge Q{idx + 1}"
    if qa_id in processed_ids:
        print(f"⏭️ Skipping already processed: {qa_id}")
        continue

    context_text = row["context"]
    question = row["question"]

    query_full = (
        f"{question}\n"
        "Read the question and state your answer. "
        "State your answer, starting with 'Answer:', ending with two line breaks.\n\n"
    )

    try:
        cc = ContextCiter(
            model,
            tokenizer,
            context=context_text,
            query=query_full,
            generate_kwargs={
                "max_new_tokens": 2048,
                "do_sample": False,
                "pad_token_id": tokenizer.eos_token_id,
            }
        )

        raw_response = cc.response.strip()
        match = re.search(r"Answer:\s*([A-J])", raw_response)
        extracted_answer = match.group(1).strip() if match else None

        result = cc.get_attributions(as_dataframe=True)
        if isinstance(result, pd.io.formats.style.Styler):
            result = result.data

        if isinstance(result, pd.DataFrame):
            result["QA_ID"] = qa_id
            result["Extracted_Answer"] = extracted_answer
            filename = f"{qa_id}.csv".replace(" ", "_")
            result.to_csv(os.path.join(output_dir, filename), index=False)
            print(f"✅ Saved attribution for {qa_id}")
        else:
            print(f"⚠️ Unexpected result type at row {idx}: {type(result)}")

    except Exception as e:
        print(f"❌ Error on row {idx} ({qa_id}): {e}")
```
