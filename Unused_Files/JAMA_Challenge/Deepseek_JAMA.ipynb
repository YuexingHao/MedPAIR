## Show Performance on Dataset 

```python
import pandas as pd

df = pd.read_csv("jama_formatted_questions.csv")
df.head()
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

model_name = "deepseek-ai/DeepSeek-R1-Distill-Qwen-14B"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name, device_map="auto", torch_dtype=torch.float16)
model.eval()
```

```python
# Ensure prompts exist in a column, say 'prompt'
prompts = df['prompt'].iloc[:5]

for i, prompt in enumerate(prompts):
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    with torch.no_grad():
        outputs = model.generate(**inputs, max_new_tokens=200, do_sample=False)
    decoded_output = tokenizer.decode(outputs[0], skip_special_tokens=True)
    print(f"\n--- Sample {i+1} ---")
    print(f"Prompt:\n{prompt}")
    print(f"\nGenerated Response:\n{decoded_output}")
```

```python
answers = df['answer'].iloc[:5]

for i in range(5):
    print(f"\n--- Sample {i+1} ---")
    print(f"Reference Answer:\n{answers[i]}")
```
