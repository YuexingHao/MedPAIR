```python
import openai
from pizza import LLMAttributor
from pizza.metrics import ProbDiffMetric
from pizza.strategies import IterativePerturbation
import os

# Set your OpenAI API key
openai.# Replace with your actual API key

# QA prompt
prompt = "Question: What is the capital of France?\nAnswer:"
reference = "The capital of France is Paris."

# Attribution setup
strategy = IterativePerturbation()
metric = ProbDiffMetric(target_completion=reference)

attributor = LLMAttributor.from_openai(
    model="gpt-3.5-turbo-instruct",  # supports logprobs
    strategy=strategy,
    metric=metric,
    max_tokens=20,
    temperature=0.0,
    top_logprobs=5
)

# Compute attribution
result = attributor.attribute(prompt)

# Visualize
result.render_heatmap()
```
