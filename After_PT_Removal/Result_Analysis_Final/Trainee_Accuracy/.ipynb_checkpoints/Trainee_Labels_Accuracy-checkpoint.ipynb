```python
import pandas as pd

# Load the data
output_path = "Trainee_Accuracy_All_Labels.xlsx - Full Results 051525.csv"
df = pd.read_csv(output_path)
df.columns

correct_answer_path = "gpt5-relevancy-combined-dec-12.csv"
correct_df = pd.read_csv(correct_answer_path)
# correct_df.columns
```

```python
# Merge the dataframes on Origin and ID_corr
merged = df.merge(
    correct_df[['ID_corr', 'answer_corr', 'data_source_corr']], 
    left_on='Origin', 
    right_on='ID_corr', 
    how='left'
)

# Add the answer_corr column to df
df['answer_corr'] = merged['answer_corr']
df['data_source_corr'] = merged['data_source_corr']

# Compare q1 with answer_corr (both converted to lowercase)
# If answer_corr is NaN (no match found), set Match to FALSE
df['Match'] = merged.apply(
    lambda row: 'TRUE' if pd.notna(row['answer_corr']) and 
                str(row['q1']).lower() == str(row['answer_corr']).lower() 
                else 'FALSE', 
    axis=1
)
```

```python
import pandas as pd
import numpy as np

# Get a random occurrence of each unique value in "Origin"
random_occurrences = df.groupby('Origin', group_keys=False).apply(
    lambda x: x.sample(n=1)  # Different random selection each run
).reset_index(drop=True)

# Calculate overall statistics
binary_responses = (random_occurrences['Match'] == 'TRUE').astype(int)
true_count = binary_responses.sum()
total_count = len(random_occurrences)
percentage = (true_count / total_count) * 100
std_dev = binary_responses.std()

print("=" * 60)
print("RANDOM OCCURRENCE ANALYSIS")
print("=" * 60)
print(f"Total unique 'Origin' values: {total_count}")
print(f"Number of TRUE responses: {true_count}")
print(f"Number of FALSE responses: {total_count - true_count}")
print(f"Percentage of TRUE: {percentage:.2f}%")
print(f"Standard Deviation: {std_dev:.4f} ({std_dev * 100:.2f}%)")
print("=" * 60)

# Analysis by data_source_corr
print("\nANALYSIS BY DATA SOURCE")
print("=" * 60)

# Remove null data_source_corr if any
random_occurrences_clean = random_occurrences[random_occurrences['data_source_corr'].notna()].copy()

# Group by data_source_corr and calculate statistics
by_source = random_occurrences_clean.groupby('data_source_corr').agg(
    Total=('Match', 'count'),
    TRUE_Count=('Match', lambda x: (x == 'TRUE').sum()),
    FALSE_Count=('Match', lambda x: (x == 'FALSE').sum())
).reset_index()

# Calculate percentage and std for each source
percentages = []
std_devs = []

for source in by_source['data_source_corr']:
    source_data = random_occurrences_clean[random_occurrences_clean['data_source_corr'] == source]['Match']
    binary_data = (source_data == 'TRUE').astype(int)
    
    pct = (binary_data.sum() / len(binary_data) * 100) if len(binary_data) > 0 else 0
    std = binary_data.std() if len(binary_data) > 1 else 0
    
    percentages.append(pct)
    std_devs.append(std)

by_source['Percentage'] = percentages
by_source['Std_Dev'] = std_devs
by_source['Std_Dev_%'] = (by_source['Std_Dev'] * 100).round(2)

# Sort by percentage descending
by_source_sorted = by_source.sort_values('Percentage', ascending=False)

print(by_source_sorted[['data_source_corr', 'Total', 'TRUE_Count', 'FALSE_Count', 
                         'Percentage', 'Std_Dev_%']].to_string(index=False))
print("=" * 60)

# Optional: Show breakdown by Origin (kept from original)
print("\nBREAKDOWN BY ORIGIN")
print("=" * 60)
breakdown = random_occurrences.groupby('Origin')['Match'].apply(
    lambda x: (x == 'TRUE').sum()
).reset_index()
breakdown.columns = ['Origin', 'TRUE_Count']
breakdown['Total'] = random_occurrences.groupby('Origin').size().values
breakdown['Percentage'] = (breakdown['TRUE_Count'] / breakdown['Total'] * 100).round(2)
print(breakdown.to_string(index=False))
print("=" * 60)

# Save results - Updated summary
summary_df = pd.DataFrame({
    'Metric': ['Total Unique Origins', 'TRUE Responses', 'FALSE Responses', 
               'Percentage TRUE', 'Standard Deviation'],
    'Value': [total_count, true_count, total_count - true_count, 
              f"{percentage:.2f}%", f"{std_dev:.4f} ({std_dev * 100:.2f}%)"]
})

# Save by_source results
by_source_summary = by_source_sorted[['data_source_corr', 'Total', 'TRUE_Count', 
                                       'FALSE_Count', 'Percentage', 'Std_Dev_%']].copy()
by_source_summary['Percentage'] = by_source_summary['Percentage'].round(2)

print("\n" + "=" * 60)
print("SUMMARY BY DATA SOURCE")
print("=" * 60)
print(f"Number of data sources: {len(by_source)}")
print(f"Highest accuracy: {by_source_sorted.iloc[0]['data_source_corr']} ({by_source_sorted.iloc[0]['Percentage']:.2f}%)")
print(f"Lowest accuracy: {by_source_sorted.iloc[-1]['data_source_corr']} ({by_source_sorted.iloc[-1]['Percentage']:.2f}%)")
print(f"Mean accuracy across sources: {by_source['Percentage'].mean():.2f}%")
print("=" * 60)
```
