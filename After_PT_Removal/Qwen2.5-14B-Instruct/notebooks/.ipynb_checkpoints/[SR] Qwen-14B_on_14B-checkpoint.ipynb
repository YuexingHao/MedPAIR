```python
from transformers import AutoModelForCausalLM, AutoTokenizer

model_name = "Qwen/Qwen2.5-14B-Instruct"

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype="auto",
    device_map="auto"
)
tokenizer = AutoTokenizer.from_pretrained(model_name)

prompt = "Give me a short introduction to large language model."
messages = [
    {"role": "system", "content": "You are Qwen, created by Alibaba Cloud. You are a helpful assistant."},
    {"role": "user", "content": prompt}
]
text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)
model_inputs = tokenizer([text], return_tensors="pt").to(model.device)

generated_ids = model.generate(
    **model_inputs,
    max_new_tokens=2048,
    do_sample=False
)
generated_ids = [
    output_ids[len(input_ids):] for input_ids, output_ids in zip(model_inputs.input_ids, generated_ids)
]

response = tokenizer.batch_decode(generated_ids, skip_special_tokens=True)[0]
```

```python
import pandas as pd 
import re
import torch

# Load data
df = pd.read_csv(paths.DATA / "Qwen72B_annotated_MedPAIR_relevancy.csv")
print("Columns in dataset:")
print(df.columns.tolist())

# Function to extract answer letter using multiple patterns
def extract_answer_letter(text):
    if pd.isna(text) or not text:
        return None
    
    # Try different patterns to extract the answer letter
    patterns = [
        r"Answer:\s*([A-J])",             # "Answer: A"
        r"Answer is\s*([A-J])",           # "Answer is A"
        r"answer is\s*([A-J])",           # "answer is A"
        r"The answer is\s*([A-J])",       # "The answer is A"
        r"the answer is\s*([A-J])",       # "the answer is A"
        r"Option\s*([A-J])",              # "Option A"
        r"option\s*([A-J])",              # "option A"
        r"My answer is\s*([A-J])",        # "My answer is A"
        r"(\n|^)([A-J])\.?\s*$",          # "A." or just "A" at end or newline
        r"select option\s*([A-J])",       # "select option A"
        r"I select\s*([A-J])",            # "I select A"
        r"I choose\s*([A-J])",            # "I choose A"
    ]
    
    for pattern in patterns:
        match = re.search(pattern, text)
        if match:
            # Some patterns have the letter in group 1, others in group 2
            return match.group(1) if len(match.groups()) == 1 else match.group(2)
    
    # If no match found, check if there's a single letter at the end
    words = text.strip().split()
    if words and len(words[-1]) == 1 and words[-1].isalpha() and words[-1].upper() in "ABCDEFGHIJ":
        return words[-1].upper()
    
    return None

# Create results dataframe
results = []

# Loop through the dataset
total_rows = len(df)
print(f"Processing {total_rows} rows...")

for idx, row in df.head(total_rows).iterrows():
    print(f"Processing row {idx+1}/{total_rows}...")
    try:
        context_text = row["Qwen72B_High_Relevance"]
        question = row["question_options_x"]
        
        # Improved prompt with clearer instructions
        query_full = (
            "You are a clinical reasoning assistant. You will receive a patient case summary "
            "and a multiple-choice question.\n\n"
            f"{context_text}\n\n"
            f"{question}\n\n"
            "Please select the single most appropriate answer. Respond only in the following format:\n\n"
            "Answer: <LETTER>"
    )
    
        # Generate prediction using the model
        inputs = tokenizer(query_full, return_tensors="pt").to(model.device)
        with torch.no_grad():
            output = model.generate(
                **inputs,
                max_new_tokens=512,
                do_sample=False,
                pad_token_id=tokenizer.eos_token_id
        )

        # Decode the generated response
        raw_response = tokenizer.decode(output[0][inputs.input_ids.shape[1]:], skip_special_tokens=True).strip()
        
        # Extract the answer letter using improved function
        extracted_answer = extract_answer_letter(raw_response)
        
        # If still no answer found, log more details for debugging
        if extracted_answer is None:
            print(f"⚠️ Could not extract answer from response for row {idx+1}:")
            print(f"Response: {raw_response[:100]}...")
        
        # Create result entry
        qa_id = f"Merge Q{idx + 1}"
        result_entry = {
            "QA_ID": qa_id,
            "Origin": row.get("Origin", ""),
            "data_source": row.get("data_source_corr", ""),
            "Raw_Response": raw_response,
            "Extracted_Answer": extracted_answer
        }
        
        results.append(result_entry)
        print(f"✅ Processed {qa_id}: Answer = {extracted_answer}")
        
        # Save progress every 10 items (increased frequency for safety)
        if (idx + 1) % 10 == 0:
            temp_df = pd.DataFrame(results)
            temp_df.to_csv(paths.PREDICTIONS / "[SR]_Qwen_14B_predictions_72Bprogress.csv", index=False)
            print(f"Saved progress to CSV after {idx+1} items")

    except Exception as e:
        print(f"❌ Error on row {idx}: {str(e)}")
        # Still try to save the entry with error info
        qa_id = f"Merge Q{idx + 1}"
        results.append({
            "QA_ID": qa_id,
            "Origin": row.get("Origin", ""),
            "data_source": row.get("data_source_corr", ""),
            "Raw_Response": f"ERROR: {str(e)}",
            "Extracted_Answer": None
        })

# Save final results
output_df = pd.DataFrame(results)
output_file = paths.PREDICTIONS / "[SR]_Qwen_14B_predictions_Qwen72B.csv"
output_df.to_csv(output_file, index=False)
print(f"Saved all predictions to {output_file}")
```

```python
import pandas as pd
import re

# Load your original data to get the metadata
df_original = pd.read_csv(paths.DATA / "After_Removal_High_qwen_72B_predictions.csv")

# Paste your console output as a multi-line string here
console_output = """
Processing row 1/1297...
✅ Processed Merge Q12: Answer = A
Processing row 2/1297...
✅ Processed Merge Q12: Answer = C
Processing row 3/1297...
✅ Processed Merge Q12: Answer = D
Processing row 4/1297...
✅ Processed Merge Q12: Answer = A
Processing row 5/1297...
✅ Processed Merge Q12: Answer = D
Processing row 6/1297...
✅ Processed Merge Q12: Answer = C
Processing row 7/1297...
✅ Processed Merge Q12: Answer = C
Processing row 8/1297...
✅ Processed Merge Q12: Answer = B
Processing row 9/1297...
✅ Processed Merge Q12: Answer = B
Processing row 10/1297...
✅ Processed Merge Q12: Answer = D
Saved progress to CSV after 10 items
Processing row 11/1297...
✅ Processed Merge Q12: Answer = D
Processing row 12/1297...
✅ Processed Merge Q12: Answer = B
Processing row 13/1297...
✅ Processed Merge Q12: Answer = B
Processing row 14/1297...
✅ Processed Merge Q12: Answer = A
Processing row 15/1297...
✅ Processed Merge Q12: Answer = C
Processing row 16/1297...
✅ Processed Merge Q12: Answer = C
Processing row 17/1297...
✅ Processed Merge Q12: Answer = D
Processing row 18/1297...
✅ Processed Merge Q12: Answer = C
Processing row 19/1297...
✅ Processed Merge Q12: Answer = A
Processing row 20/1297...
✅ Processed Merge Q12: Answer = D
Saved progress to CSV after 20 items
Processing row 21/1297...
✅ Processed Merge Q12: Answer = D
Processing row 22/1297...
✅ Processed Merge Q12: Answer = D
Processing row 23/1297...
✅ Processed Merge Q12: Answer = B
Processing row 24/1297...
✅ Processed Merge Q12: Answer = C
Processing row 25/1297...
✅ Processed Merge Q12: Answer = D
Processing row 26/1297...
✅ Processed Merge Q12: Answer = C
Processing row 27/1297...
✅ Processed Merge Q12: Answer = B
Processing row 28/1297...
✅ Processed Merge Q12: Answer = C
Processing row 29/1297...
✅ Processed Merge Q12: Answer = C
Processing row 30/1297...
✅ Processed Merge Q12: Answer = D
Saved progress to CSV after 30 items
Processing row 31/1297...
✅ Processed Merge Q12: Answer = B
Processing row 32/1297...
✅ Processed Merge Q12: Answer = D
Processing row 33/1297...
✅ Processed Merge Q12: Answer = B
Processing row 34/1297...
✅ Processed Merge Q12: Answer = C
Processing row 35/1297...
✅ Processed Merge Q12: Answer = A
Processing row 36/1297...
✅ Processed Merge Q12: Answer = B
Processing row 37/1297...
✅ Processed Merge Q12: Answer = C
Processing row 38/1297...
✅ Processed Merge Q12: Answer = A
Processing row 39/1297...
✅ Processed Merge Q12: Answer = D
Processing row 40/1297...
✅ Processed Merge Q12: Answer = A
Saved progress to CSV after 40 items
Processing row 41/1297...
✅ Processed Merge Q12: Answer = C
Processing row 42/1297...
✅ Processed Merge Q12: Answer = B
Processing row 43/1297...
✅ Processed Merge Q12: Answer = D
Processing row 44/1297...
✅ Processed Merge Q12: Answer = C
Processing row 45/1297...
✅ Processed Merge Q12: Answer = B
Processing row 46/1297...
✅ Processed Merge Q12: Answer = C
Processing row 47/1297...
✅ Processed Merge Q12: Answer = B
Processing row 48/1297...
✅ Processed Merge Q12: Answer = A
Processing row 49/1297...
✅ Processed Merge Q12: Answer = A
Processing row 50/1297...
✅ Processed Merge Q12: Answer = B
Saved progress to CSV after 50 items
Processing row 51/1297...
✅ Processed Merge Q12: Answer = C
Processing row 52/1297...
✅ Processed Merge Q12: Answer = D
Processing row 53/1297...
✅ Processed Merge Q12: Answer = D
Processing row 54/1297...
✅ Processed Merge Q12: Answer = B
Processing row 55/1297...
✅ Processed Merge Q12: Answer = B
Processing row 56/1297...
✅ Processed Merge Q12: Answer = D
Processing row 57/1297...
✅ Processed Merge Q12: Answer = A
Processing row 58/1297...
✅ Processed Merge Q12: Answer = A
Processing row 59/1297...
✅ Processed Merge Q12: Answer = C
Processing row 60/1297...
✅ Processed Merge Q12: Answer = C
Saved progress to CSV after 60 items
Processing row 61/1297...
✅ Processed Merge Q12: Answer = B
Processing row 62/1297...
✅ Processed Merge Q12: Answer = C
Processing row 63/1297...
✅ Processed Merge Q12: Answer = B
Processing row 64/1297...
✅ Processed Merge Q12: Answer = C
Processing row 65/1297...
✅ Processed Merge Q12: Answer = B
Processing row 66/1297...
⚠️ Could not extract answer from response for row 66:
Response: To determine the most likely underlying etiology for the patient's chief concern of muscle spasms, l...
✅ Processed Merge Q66: Answer = None
Processing row 67/1297...
✅ Processed Merge Q66: Answer = C
Processing row 68/1297...
✅ Processed Merge Q66: Answer = A
Processing row 69/1297...
✅ Processed Merge Q66: Answer = B
Processing row 70/1297...
✅ Processed Merge Q66: Answer = A
Saved progress to CSV after 70 items
Processing row 71/1297...
✅ Processed Merge Q66: Answer = B
Processing row 72/1297...
✅ Processed Merge Q66: Answer = B
Processing row 73/1297...
✅ Processed Merge Q66: Answer = C
Processing row 74/1297...
✅ Processed Merge Q66: Answer = C
Processing row 75/1297...
✅ Processed Merge Q66: Answer = D
Processing row 76/1297...
✅ Processed Merge Q66: Answer = D
Processing row 77/1297...
✅ Processed Merge Q66: Answer = B
Processing row 78/1297...
✅ Processed Merge Q66: Answer = B
Processing row 79/1297...
✅ Processed Merge Q66: Answer = D
Processing row 80/1297...
✅ Processed Merge Q66: Answer = C
Saved progress to CSV after 80 items
Processing row 81/1297...
✅ Processed Merge Q66: Answer = C
Processing row 82/1297...
✅ Processed Merge Q66: Answer = D
Processing row 83/1297...
✅ Processed Merge Q66: Answer = A
Processing row 84/1297...
✅ Processed Merge Q66: Answer = C
Processing row 85/1297...
✅ Processed Merge Q66: Answer = D
Processing row 86/1297...
✅ Processed Merge Q66: Answer = D
Processing row 87/1297...
✅ Processed Merge Q66: Answer = B
Processing row 88/1297...
✅ Processed Merge Q66: Answer = B
Processing row 89/1297...
✅ Processed Merge Q66: Answer = B
Processing row 90/1297...
✅ Processed Merge Q66: Answer = A
Saved progress to CSV after 90 items
Processing row 91/1297...
✅ Processed Merge Q66: Answer = A
Processing row 92/1297...
✅ Processed Merge Q66: Answer = A
Processing row 93/1297...
✅ Processed Merge Q66: Answer = D
Processing row 94/1297...
✅ Processed Merge Q66: Answer = A
Processing row 95/1297...
✅ Processed Merge Q66: Answer = B
Processing row 96/1297...
✅ Processed Merge Q66: Answer = B
Processing row 97/1297...
✅ Processed Merge Q66: Answer = C
Processing row 98/1297...
✅ Processed Merge Q66: Answer = B
Processing row 99/1297...
✅ Processed Merge Q66: Answer = D
Processing row 100/1297...
✅ Processed Merge Q66: Answer = C
Saved progress to CSV after 100 items
Processing row 101/1297...
✅ Processed Merge Q66: Answer = D
Processing row 102/1297...
✅ Processed Merge Q66: Answer = C
Processing row 103/1297...
✅ Processed Merge Q66: Answer = C
Processing row 104/1297...
✅ Processed Merge Q66: Answer = A
Processing row 105/1297...
✅ Processed Merge Q66: Answer = D
Processing row 106/1297...
✅ Processed Merge Q66: Answer = A
Processing row 107/1297...
✅ Processed Merge Q66: Answer = C
Processing row 108/1297...
✅ Processed Merge Q66: Answer = D
Processing row 109/1297...
✅ Processed Merge Q66: Answer = D
Processing row 110/1297...
✅ Processed Merge Q66: Answer = B
Saved progress to CSV after 110 items
Processing row 111/1297...
✅ Processed Merge Q66: Answer = A
Processing row 112/1297...
✅ Processed Merge Q66: Answer = D
Processing row 113/1297...
✅ Processed Merge Q66: Answer = B
Processing row 114/1297...
✅ Processed Merge Q66: Answer = B
Processing row 115/1297...
✅ Processed Merge Q66: Answer = B
Processing row 116/1297...
✅ Processed Merge Q66: Answer = A
Processing row 117/1297...
✅ Processed Merge Q66: Answer = D
Processing row 118/1297...
✅ Processed Merge Q66: Answer = A
Processing row 119/1297...
✅ Processed Merge Q66: Answer = A
Processing row 120/1297...
✅ Processed Merge Q66: Answer = A
Saved progress to CSV after 120 items
Processing row 121/1297...
✅ Processed Merge Q66: Answer = B
Processing row 122/1297...
✅ Processed Merge Q66: Answer = A
Processing row 123/1297...
✅ Processed Merge Q66: Answer = C
Processing row 124/1297...
✅ Processed Merge Q66: Answer = B
Processing row 125/1297...
✅ Processed Merge Q66: Answer = C
Processing row 126/1297...
✅ Processed Merge Q66: Answer = D
Processing row 127/1297...
✅ Processed Merge Q66: Answer = C
Processing row 128/1297...
✅ Processed Merge Q66: Answer = D
Processing row 129/1297...
✅ Processed Merge Q66: Answer = D
Processing row 130/1297...
✅ Processed Merge Q66: Answer = C
Saved progress to CSV after 130 items
Processing row 131/1297...
✅ Processed Merge Q66: Answer = D
Processing row 132/1297...
✅ Processed Merge Q66: Answer = D
Processing row 133/1297...
✅ Processed Merge Q66: Answer = C
Processing row 134/1297...
✅ Processed Merge Q66: Answer = C
Processing row 135/1297...
✅ Processed Merge Q66: Answer = D
Processing row 136/1297...
✅ Processed Merge Q66: Answer = B
Processing row 137/1297...
✅ Processed Merge Q66: Answer = C
Processing row 138/1297...
✅ Processed Merge Q66: Answer = A
Processing row 139/1297...
✅ Processed Merge Q66: Answer = A
Processing row 140/1297...
✅ Processed Merge Q66: Answer = B
Saved progress to CSV after 140 items
Processing row 141/1297...
✅ Processed Merge Q66: Answer = A
Processing row 142/1297...
✅ Processed Merge Q66: Answer = C
Processing row 143/1297...
✅ Processed Merge Q66: Answer = B
Processing row 144/1297...
✅ Processed Merge Q66: Answer = C
Processing row 145/1297...
✅ Processed Merge Q66: Answer = C
Processing row 146/1297...
✅ Processed Merge Q66: Answer = B
Processing row 147/1297...
✅ Processed Merge Q66: Answer = D
Processing row 148/1297...
✅ Processed Merge Q66: Answer = B
Processing row 149/1297...
✅ Processed Merge Q66: Answer = C
Processing row 150/1297...
✅ Processed Merge Q66: Answer = D
Saved progress to CSV after 150 items
Processing row 151/1297...
✅ Processed Merge Q66: Answer = D
Processing row 152/1297...
✅ Processed Merge Q66: Answer = D
Processing row 153/1297...
✅ Processed Merge Q66: Answer = C
Processing row 154/1297...
✅ Processed Merge Q66: Answer = B
Processing row 155/1297...
✅ Processed Merge Q66: Answer = D
Processing row 156/1297...
✅ Processed Merge Q66: Answer = B
Processing row 157/1297...
✅ Processed Merge Q66: Answer = B
Processing row 158/1297...
✅ Processed Merge Q66: Answer = C
Processing row 159/1297...
✅ Processed Merge Q66: Answer = B
Processing row 160/1297...
✅ Processed Merge Q66: Answer = A
Saved progress to CSV after 160 items
Processing row 161/1297...
✅ Processed Merge Q66: Answer = C
Processing row 162/1297...
✅ Processed Merge Q66: Answer = B
Processing row 163/1297...
✅ Processed Merge Q66: Answer = B
Processing row 164/1297...
✅ Processed Merge Q66: Answer = B
Processing row 165/1297...
✅ Processed Merge Q66: Answer = A
Processing row 166/1297...
✅ Processed Merge Q66: Answer = A
Processing row 167/1297...
✅ Processed Merge Q66: Answer = B
Processing row 168/1297...
✅ Processed Merge Q66: Answer = B
Processing row 169/1297...
✅ Processed Merge Q66: Answer = C
Processing row 170/1297...
✅ Processed Merge Q66: Answer = C
Saved progress to CSV after 170 items
Processing row 171/1297...
✅ Processed Merge Q66: Answer = B
Processing row 172/1297...
✅ Processed Merge Q66: Answer = C
Processing row 173/1297...
✅ Processed Merge Q66: Answer = C
Processing row 174/1297...
✅ Processed Merge Q66: Answer = A
Processing row 175/1297...
✅ Processed Merge Q66: Answer = A
Processing row 176/1297...
✅ Processed Merge Q66: Answer = B
Processing row 177/1297...
✅ Processed Merge Q66: Answer = A
Processing row 178/1297...
✅ Processed Merge Q66: Answer = C
Processing row 179/1297...
✅ Processed Merge Q66: Answer = C
Processing row 180/1297...
✅ Processed Merge Q66: Answer = D
Saved progress to CSV after 180 items
Processing row 181/1297...
✅ Processed Merge Q66: Answer = D
Processing row 182/1297...
✅ Processed Merge Q66: Answer = B
Processing row 183/1297...
✅ Processed Merge Q66: Answer = B
Processing row 184/1297...
✅ Processed Merge Q66: Answer = D
Processing row 185/1297...
✅ Processed Merge Q66: Answer = D
Processing row 186/1297...
✅ Processed Merge Q66: Answer = C
Processing row 187/1297...
✅ Processed Merge Q66: Answer = B
Processing row 188/1297...
✅ Processed Merge Q66: Answer = B
Processing row 189/1297...
✅ Processed Merge Q66: Answer = C
Processing row 190/1297...
✅ Processed Merge Q66: Answer = D
Saved progress to CSV after 190 items
Processing row 191/1297...
✅ Processed Merge Q66: Answer = D
Processing row 192/1297...
✅ Processed Merge Q66: Answer = B
Processing row 193/1297...
✅ Processed Merge Q66: Answer = C
Processing row 194/1297...
✅ Processed Merge Q66: Answer = A
Processing row 195/1297...
✅ Processed Merge Q66: Answer = C
Processing row 196/1297...
✅ Processed Merge Q66: Answer = C
Processing row 197/1297...
✅ Processed Merge Q66: Answer = D
Processing row 198/1297...
✅ Processed Merge Q66: Answer = B
Processing row 199/1297...
✅ Processed Merge Q66: Answer = C
Processing row 200/1297...
✅ Processed Merge Q66: Answer = C
Saved progress to CSV after 200 items
Processing row 201/1297...
✅ Processed Merge Q66: Answer = D
Processing row 202/1297...
✅ Processed Merge Q66: Answer = B
Processing row 203/1297...
✅ Processed Merge Q66: Answer = A
Processing row 204/1297...
✅ Processed Merge Q66: Answer = C
Processing row 205/1297...
✅ Processed Merge Q66: Answer = A
Processing row 206/1297...
✅ Processed Merge Q66: Answer = C
Processing row 207/1297...
✅ Processed Merge Q66: Answer = B
Processing row 208/1297...
✅ Processed Merge Q66: Answer = A
Processing row 209/1297...
✅ Processed Merge Q66: Answer = D
Processing row 210/1297...
✅ Processed Merge Q66: Answer = B
Saved progress to CSV after 210 items
Processing row 211/1297...
✅ Processed Merge Q66: Answer = C
Processing row 212/1297...
✅ Processed Merge Q66: Answer = C
Processing row 213/1297...
✅ Processed Merge Q66: Answer = B
Processing row 214/1297...
✅ Processed Merge Q66: Answer = A
Processing row 215/1297...
✅ Processed Merge Q66: Answer = D
Processing row 216/1297...
✅ Processed Merge Q66: Answer = A
Processing row 217/1297...
✅ Processed Merge Q66: Answer = B
Processing row 218/1297...
✅ Processed Merge Q66: Answer = C
Processing row 219/1297...
✅ Processed Merge Q66: Answer = A
Processing row 220/1297...
✅ Processed Merge Q66: Answer = D
Saved progress to CSV after 220 items
Processing row 221/1297...
✅ Processed Merge Q66: Answer = C
Processing row 222/1297...
✅ Processed Merge Q66: Answer = C
Processing row 223/1297...
✅ Processed Merge Q66: Answer = A
Processing row 224/1297...
✅ Processed Merge Q66: Answer = B
Processing row 225/1297...
✅ Processed Merge Q66: Answer = D
Processing row 226/1297...
✅ Processed Merge Q66: Answer = C
Processing row 227/1297...
✅ Processed Merge Q66: Answer = C
Processing row 228/1297...
✅ Processed Merge Q66: Answer = C
Processing row 229/1297...
✅ Processed Merge Q66: Answer = C
Processing row 230/1297...
✅ Processed Merge Q66: Answer = C
Saved progress to CSV after 230 items
Processing row 231/1297...
✅ Processed Merge Q66: Answer = B
Processing row 232/1297...
✅ Processed Merge Q66: Answer = B
Processing row 233/1297...
✅ Processed Merge Q66: Answer = D
Processing row 234/1297...
✅ Processed Merge Q66: Answer = B
Processing row 235/1297...
✅ Processed Merge Q66: Answer = C
Processing row 236/1297...
✅ Processed Merge Q66: Answer = A
Processing row 237/1297...
✅ Processed Merge Q66: Answer = A
Processing row 238/1297...
✅ Processed Merge Q66: Answer = A
Processing row 239/1297...
✅ Processed Merge Q66: Answer = B
Processing row 240/1297...
✅ Processed Merge Q66: Answer = A
Saved progress to CSV after 240 items
Processing row 241/1297...
✅ Processed Merge Q66: Answer = C
Processing row 242/1297...
✅ Processed Merge Q66: Answer = C
Processing row 243/1297...
✅ Processed Merge Q66: Answer = D
Processing row 244/1297...
✅ Processed Merge Q66: Answer = B
Processing row 245/1297...
✅ Processed Merge Q66: Answer = B
Processing row 246/1297...
✅ Processed Merge Q66: Answer = B
Processing row 247/1297...
✅ Processed Merge Q66: Answer = C
Processing row 248/1297...
✅ Processed Merge Q66: Answer = H
Processing row 249/1297...
✅ Processed Merge Q66: Answer = A
Processing row 250/1297...
✅ Processed Merge Q66: Answer = E
Saved progress to CSV after 250 items
Processing row 251/1297...
✅ Processed Merge Q66: Answer = D
Processing row 252/1297...
⚠️ Could not extract answer from response for row 252:
Response: To determine the most likely diagnosis, let's analyze the key points from the patient's presentation...
✅ Processed Merge Q252: Answer = None
Processing row 253/1297...
✅ Processed Merge Q252: Answer = B
Processing row 254/1297...
⚠️ Could not extract answer from response for row 254:
Response: To understand the correct answer, let's break down the key points from the patient's presentation:

...
✅ Processed Merge Q254: Answer = None
Processing row 255/1297...
✅ Processed Merge Q254: Answer = D
Processing row 256/1297...
✅ Processed Merge Q254: Answer = B
Processing row 257/1297...
✅ Processed Merge Q254: Answer = D
Processing row 258/1297...
⚠️ Could not extract answer from response for row 258:
Response: To determine the most appropriate next step in managing this patient, let's analyze the information ...
✅ Processed Merge Q258: Answer = None
Processing row 259/1297...
✅ Processed Merge Q258: Answer = B
Processing row 260/1297...
✅ Processed Merge Q258: Answer = D
Saved progress to CSV after 260 items
Processing row 261/1297...
✅ Processed Merge Q258: Answer = A
Processing row 262/1297...
✅ Processed Merge Q258: Answer = A
Processing row 263/1297...
✅ Processed Merge Q258: Answer = B
Processing row 264/1297...
✅ Processed Merge Q258: Answer = D
Processing row 265/1297...
✅ Processed Merge Q258: Answer = A
Processing row 266/1297...
✅ Processed Merge Q258: Answer = D
Processing row 267/1297...
✅ Processed Merge Q258: Answer = D
Processing row 268/1297...
✅ Processed Merge Q258: Answer = G
Processing row 269/1297...
✅ Processed Merge Q258: Answer = G
Processing row 270/1297...
✅ Processed Merge Q258: Answer = D
Saved progress to CSV after 270 items
Processing row 271/1297...
✅ Processed Merge Q258: Answer = B
Processing row 272/1297...
✅ Processed Merge Q258: Answer = B
Processing row 273/1297...
✅ Processed Merge Q258: Answer = D
Processing row 274/1297...
✅ Processed Merge Q258: Answer = G
Processing row 275/1297...
✅ Processed Merge Q258: Answer = G
Processing row 276/1297...
✅ Processed Merge Q258: Answer = B
Processing row 277/1297...
✅ Processed Merge Q258: Answer = G
Processing row 278/1297...
✅ Processed Merge Q258: Answer = B
Processing row 279/1297...
✅ Processed Merge Q258: Answer = F
Processing row 280/1297...
✅ Processed Merge Q258: Answer = J
Saved progress to CSV after 280 items
Processing row 281/1297...
✅ Processed Merge Q258: Answer = C
Processing row 282/1297...
✅ Processed Merge Q258: Answer = C
Processing row 283/1297...
✅ Processed Merge Q258: Answer = I
Processing row 284/1297...
✅ Processed Merge Q258: Answer = G
Processing row 285/1297...
✅ Processed Merge Q258: Answer = C
Processing row 286/1297...
✅ Processed Merge Q258: Answer = C
Processing row 287/1297...
✅ Processed Merge Q258: Answer = B
Processing row 288/1297...
✅ Processed Merge Q258: Answer = B
Processing row 289/1297...
✅ Processed Merge Q258: Answer = B
Processing row 290/1297...
✅ Processed Merge Q258: Answer = C
Saved progress to CSV after 290 items
Processing row 291/1297...
✅ Processed Merge Q258: Answer = C
Processing row 292/1297...
✅ Processed Merge Q258: Answer = B
Processing row 293/1297...
✅ Processed Merge Q258: Answer = D
Processing row 294/1297...
✅ Processed Merge Q258: Answer = C
Processing row 295/1297...
✅ Processed Merge Q258: Answer = F
Processing row 296/1297...
✅ Processed Merge Q258: Answer = G
Processing row 297/1297...
✅ Processed Merge Q258: Answer = J
Processing row 298/1297...
✅ Processed Merge Q258: Answer = A
Processing row 299/1297...
✅ Processed Merge Q258: Answer = I
Processing row 300/1297...
✅ Processed Merge Q258: Answer = F
Saved progress to CSV after 300 items
Processing row 301/1297...
✅ Processed Merge Q258: Answer = B
Processing row 302/1297...
✅ Processed Merge Q258: Answer = D
Processing row 303/1297...
✅ Processed Merge Q258: Answer = F
Processing row 304/1297...
✅ Processed Merge Q258: Answer = A
Processing row 305/1297...
✅ Processed Merge Q258: Answer = B
Processing row 306/1297...
✅ Processed Merge Q258: Answer = H
Processing row 307/1297...
✅ Processed Merge Q258: Answer = C
Processing row 308/1297...
✅ Processed Merge Q258: Answer = D
Processing row 309/1297...
✅ Processed Merge Q258: Answer = F
Processing row 310/1297...
✅ Processed Merge Q258: Answer = G
Saved progress to CSV after 310 items
Processing row 311/1297...
✅ Processed Merge Q258: Answer = A
Processing row 312/1297...
✅ Processed Merge Q258: Answer = H
Processing row 313/1297...
✅ Processed Merge Q258: Answer = A
Processing row 314/1297...
✅ Processed Merge Q258: Answer = E
Processing row 315/1297...
✅ Processed Merge Q258: Answer = H
Processing row 316/1297...
✅ Processed Merge Q258: Answer = B
Processing row 317/1297...
✅ Processed Merge Q258: Answer = I
Processing row 318/1297...
✅ Processed Merge Q258: Answer = C
Processing row 319/1297...
✅ Processed Merge Q258: Answer = J
Processing row 320/1297...
✅ Processed Merge Q258: Answer = G
Saved progress to CSV after 320 items
Processing row 321/1297...
✅ Processed Merge Q258: Answer = B
Processing row 322/1297...
✅ Processed Merge Q258: Answer = G
Processing row 323/1297...
✅ Processed Merge Q258: Answer = A
Processing row 324/1297...
✅ Processed Merge Q258: Answer = B
Processing row 325/1297...
✅ Processed Merge Q258: Answer = A
Processing row 326/1297...
✅ Processed Merge Q258: Answer = J
Processing row 327/1297...
⚠️ Could not extract answer from response for row 327:
Response: To determine the most appropriate next step in managing a 16-month-old child who has fallen from a s...
✅ Processed Merge Q327: Answer = None
Processing row 328/1297...
✅ Processed Merge Q327: Answer = E
Processing row 329/1297...
✅ Processed Merge Q327: Answer = C
Processing row 330/1297...
✅ Processed Merge Q327: Answer = C
Saved progress to CSV after 330 items
Processing row 331/1297...
✅ Processed Merge Q327: Answer = A
Processing row 332/1297...
✅ Processed Merge Q327: Answer = E
Processing row 333/1297...
✅ Processed Merge Q327: Answer = C
Processing row 334/1297...
✅ Processed Merge Q327: Answer = A
Processing row 335/1297...
✅ Processed Merge Q327: Answer = B
Processing row 336/1297...
✅ Processed Merge Q327: Answer = C
Processing row 337/1297...
✅ Processed Merge Q327: Answer = A
Processing row 338/1297...
✅ Processed Merge Q327: Answer = G
Processing row 339/1297...
✅ Processed Merge Q327: Answer = B
Processing row 340/1297...
✅ Processed Merge Q327: Answer = F
Saved progress to CSV after 340 items
Processing row 341/1297...
✅ Processed Merge Q327: Answer = E
Processing row 342/1297...
✅ Processed Merge Q327: Answer = J
Processing row 343/1297...
✅ Processed Merge Q327: Answer = D
Processing row 344/1297...
✅ Processed Merge Q327: Answer = B
Processing row 345/1297...
✅ Processed Merge Q327: Answer = D
Processing row 346/1297...
✅ Processed Merge Q327: Answer = B
Processing row 347/1297...
✅ Processed Merge Q327: Answer = H
Processing row 348/1297...
✅ Processed Merge Q327: Answer = B
Processing row 349/1297...
✅ Processed Merge Q327: Answer = E
Processing row 350/1297...
✅ Processed Merge Q327: Answer = D
Saved progress to CSV after 350 items
Processing row 351/1297...
✅ Processed Merge Q327: Answer = B
Processing row 352/1297...
✅ Processed Merge Q327: Answer = J
Processing row 353/1297...
✅ Processed Merge Q327: Answer = B
Processing row 354/1297...
✅ Processed Merge Q327: Answer = A
Processing row 355/1297...
⚠️ Could not extract answer from response for row 355:
Response: To determine the most appropriate therapeutic intervention for this patient, let's analyze her situa...
✅ Processed Merge Q355: Answer = None
Processing row 356/1297...
✅ Processed Merge Q355: Answer = F
Processing row 357/1297...
✅ Processed Merge Q355: Answer = G
Processing row 358/1297...
✅ Processed Merge Q355: Answer = A
Processing row 359/1297...
✅ Processed Merge Q355: Answer = A
Processing row 360/1297...
✅ Processed Merge Q355: Answer = A
Saved progress to CSV after 360 items
Processing row 361/1297...
✅ Processed Merge Q355: Answer = I
Processing row 362/1297...
✅ Processed Merge Q355: Answer = F
Processing row 363/1297...
⚠️ Could not extract answer from response for row 363:
Response: To determine the most appropriate management for this patient, let's analyze the information provide...
✅ Processed Merge Q363: Answer = None
Processing row 364/1297...
✅ Processed Merge Q363: Answer = D
Processing row 365/1297...
✅ Processed Merge Q363: Answer = A
Processing row 366/1297...
✅ Processed Merge Q363: Answer = A
Processing row 367/1297...
✅ Processed Merge Q363: Answer = D
Processing row 368/1297...
✅ Processed Merge Q363: Answer = E
Processing row 369/1297...
✅ Processed Merge Q363: Answer = H
Processing row 370/1297...
✅ Processed Merge Q363: Answer = F
Saved progress to CSV after 370 items
Processing row 371/1297...
✅ Processed Merge Q363: Answer = C
Processing row 372/1297...
✅ Processed Merge Q363: Answer = I
Processing row 373/1297...
✅ Processed Merge Q363: Answer = I
Processing row 374/1297...
✅ Processed Merge Q363: Answer = D
Processing row 375/1297...
✅ Processed Merge Q363: Answer = I
Processing row 376/1297...
✅ Processed Merge Q363: Answer = I
Processing row 377/1297...
✅ Processed Merge Q363: Answer = E
Processing row 378/1297...
✅ Processed Merge Q363: Answer = C
Processing row 379/1297...
✅ Processed Merge Q363: Answer = J
Processing row 380/1297...
✅ Processed Merge Q363: Answer = F
Saved progress to CSV after 380 items
Processing row 381/1297...
✅ Processed Merge Q363: Answer = F
Processing row 382/1297...
✅ Processed Merge Q363: Answer = C
Processing row 383/1297...
✅ Processed Merge Q363: Answer = D
Processing row 384/1297...
✅ Processed Merge Q363: Answer = D
Processing row 385/1297...
✅ Processed Merge Q363: Answer = D
Processing row 386/1297...
✅ Processed Merge Q363: Answer = C
Processing row 387/1297...
✅ Processed Merge Q363: Answer = D
Processing row 388/1297...
✅ Processed Merge Q363: Answer = I
Processing row 389/1297...
✅ Processed Merge Q363: Answer = C
Processing row 390/1297...
✅ Processed Merge Q363: Answer = I
Saved progress to CSV after 390 items
Processing row 391/1297...
✅ Processed Merge Q363: Answer = J
Processing row 392/1297...
✅ Processed Merge Q363: Answer = I
Processing row 393/1297...
✅ Processed Merge Q363: Answer = H
Processing row 394/1297...
✅ Processed Merge Q363: Answer = C
Processing row 395/1297...
✅ Processed Merge Q363: Answer = H
Processing row 396/1297...
✅ Processed Merge Q363: Answer = B
Processing row 397/1297...
✅ Processed Merge Q363: Answer = B
Processing row 398/1297...
✅ Processed Merge Q363: Answer = G
Processing row 399/1297...
✅ Processed Merge Q363: Answer = J
Processing row 400/1297...
✅ Processed Merge Q363: Answer = C
Saved progress to CSV after 400 items
Processing row 401/1297...
✅ Processed Merge Q363: Answer = G
Processing row 402/1297...
✅ Processed Merge Q363: Answer = D
Processing row 403/1297...
✅ Processed Merge Q363: Answer = A
Processing row 404/1297...
✅ Processed Merge Q363: Answer = F
Processing row 405/1297...
✅ Processed Merge Q363: Answer = C
Processing row 406/1297...
✅ Processed Merge Q363: Answer = C
Processing row 407/1297...
✅ Processed Merge Q363: Answer = A
Processing row 408/1297...
✅ Processed Merge Q363: Answer = B
Processing row 409/1297...
✅ Processed Merge Q363: Answer = B
Processing row 410/1297...
✅ Processed Merge Q363: Answer = C
Saved progress to CSV after 410 items
Processing row 411/1297...
✅ Processed Merge Q363: Answer = E
Processing row 412/1297...
✅ Processed Merge Q363: Answer = B
Processing row 413/1297...
✅ Processed Merge Q363: Answer = D
Processing row 414/1297...
✅ Processed Merge Q363: Answer = C
Processing row 415/1297...
✅ Processed Merge Q363: Answer = E
Processing row 416/1297...
✅ Processed Merge Q363: Answer = E
Processing row 417/1297...
✅ Processed Merge Q363: Answer = A
Processing row 418/1297...
✅ Processed Merge Q363: Answer = G
Processing row 419/1297...
✅ Processed Merge Q363: Answer = C
Processing row 420/1297...
✅ Processed Merge Q363: Answer = J
Saved progress to CSV after 420 items
Processing row 421/1297...
✅ Processed Merge Q363: Answer = A
Processing row 422/1297...
✅ Processed Merge Q363: Answer = A
Processing row 423/1297...
✅ Processed Merge Q363: Answer = A
Processing row 424/1297...
✅ Processed Merge Q363: Answer = E
Processing row 425/1297...
✅ Processed Merge Q363: Answer = D
Processing row 426/1297...
✅ Processed Merge Q363: Answer = C
Processing row 427/1297...
✅ Processed Merge Q363: Answer = A
Processing row 428/1297...
✅ Processed Merge Q363: Answer = B
Processing row 429/1297...
✅ Processed Merge Q363: Answer = E
Processing row 430/1297...
✅ Processed Merge Q363: Answer = A
Saved progress to CSV after 430 items
Processing row 431/1297...
✅ Processed Merge Q363: Answer = D
Processing row 432/1297...
✅ Processed Merge Q363: Answer = H
Processing row 433/1297...
✅ Processed Merge Q363: Answer = J
Processing row 434/1297...
✅ Processed Merge Q363: Answer = D
Processing row 435/1297...
✅ Processed Merge Q363: Answer = J
Processing row 436/1297...
✅ Processed Merge Q363: Answer = C
Processing row 437/1297...
✅ Processed Merge Q363: Answer = I
Processing row 438/1297...
✅ Processed Merge Q363: Answer = A
Processing row 439/1297...
✅ Processed Merge Q363: Answer = G
Processing row 440/1297...
✅ Processed Merge Q363: Answer = C
Saved progress to CSV after 440 items
Processing row 441/1297...
✅ Processed Merge Q363: Answer = B
Processing row 442/1297...
✅ Processed Merge Q363: Answer = C
Processing row 443/1297...
✅ Processed Merge Q363: Answer = D
Processing row 444/1297...
✅ Processed Merge Q363: Answer = I
Processing row 445/1297...
✅ Processed Merge Q363: Answer = G
Processing row 446/1297...
✅ Processed Merge Q363: Answer = D
Processing row 447/1297...
✅ Processed Merge Q363: Answer = B
Processing row 448/1297...
✅ Processed Merge Q363: Answer = A
Processing row 449/1297...
✅ Processed Merge Q363: Answer = A
Processing row 450/1297...
✅ Processed Merge Q363: Answer = J
Saved progress to CSV after 450 items
Processing row 451/1297...
✅ Processed Merge Q363: Answer = E
Processing row 452/1297...
✅ Processed Merge Q363: Answer = B
Processing row 453/1297...
✅ Processed Merge Q363: Answer = C
Processing row 454/1297...
✅ Processed Merge Q363: Answer = F
Processing row 455/1297...
✅ Processed Merge Q363: Answer = E
Processing row 456/1297...
✅ Processed Merge Q363: Answer = C
Processing row 457/1297...
✅ Processed Merge Q363: Answer = C
Processing row 458/1297...
✅ Processed Merge Q363: Answer = C
Processing row 459/1297...
✅ Processed Merge Q363: Answer = H
Processing row 460/1297...
✅ Processed Merge Q363: Answer = B
Saved progress to CSV after 460 items
Processing row 461/1297...
✅ Processed Merge Q363: Answer = B
Processing row 462/1297...
✅ Processed Merge Q363: Answer = D
Processing row 463/1297...
✅ Processed Merge Q363: Answer = A
Processing row 464/1297...
✅ Processed Merge Q363: Answer = J
Processing row 465/1297...
✅ Processed Merge Q363: Answer = G
Processing row 466/1297...
✅ Processed Merge Q363: Answer = D
Processing row 467/1297...
✅ Processed Merge Q363: Answer = B
Processing row 468/1297...
✅ Processed Merge Q363: Answer = D
Processing row 469/1297...
✅ Processed Merge Q363: Answer = B
Processing row 470/1297...
✅ Processed Merge Q363: Answer = D
Saved progress to CSV after 470 items
Processing row 471/1297...
✅ Processed Merge Q363: Answer = C
Processing row 472/1297...
✅ Processed Merge Q363: Answer = B
Processing row 473/1297...
✅ Processed Merge Q363: Answer = I
Processing row 474/1297...
✅ Processed Merge Q363: Answer = I
Processing row 475/1297...
✅ Processed Merge Q363: Answer = G
Processing row 476/1297...
✅ Processed Merge Q363: Answer = B
Processing row 477/1297...
✅ Processed Merge Q363: Answer = A
Processing row 478/1297...
✅ Processed Merge Q363: Answer = E
Processing row 479/1297...
✅ Processed Merge Q363: Answer = F
Processing row 480/1297...
✅ Processed Merge Q363: Answer = G
Saved progress to CSV after 480 items
Processing row 481/1297...
✅ Processed Merge Q363: Answer = J
Processing row 482/1297...
✅ Processed Merge Q363: Answer = C
Processing row 483/1297...
✅ Processed Merge Q363: Answer = J
Processing row 484/1297...
✅ Processed Merge Q363: Answer = G
Processing row 485/1297...
✅ Processed Merge Q363: Answer = C
Processing row 486/1297...
✅ Processed Merge Q363: Answer = I
Processing row 487/1297...
✅ Processed Merge Q363: Answer = F
Processing row 488/1297...
✅ Processed Merge Q363: Answer = A
Processing row 489/1297...
✅ Processed Merge Q363: Answer = G
Processing row 490/1297...
✅ Processed Merge Q363: Answer = I
Saved progress to CSV after 490 items
Processing row 491/1297...
✅ Processed Merge Q363: Answer = C
Processing row 492/1297...
✅ Processed Merge Q363: Answer = E
Processing row 493/1297...
✅ Processed Merge Q363: Answer = A
Processing row 494/1297...
✅ Processed Merge Q363: Answer = E
Processing row 495/1297...
✅ Processed Merge Q363: Answer = E
Processing row 496/1297...
✅ Processed Merge Q363: Answer = I
Processing row 497/1297...
✅ Processed Merge Q363: Answer = I
Processing row 498/1297...
✅ Processed Merge Q363: Answer = G
Processing row 499/1297...
✅ Processed Merge Q363: Answer = B
Processing row 500/1297...
✅ Processed Merge Q363: Answer = A
Saved progress to CSV after 500 items
Processing row 501/1297...
✅ Processed Merge Q363: Answer = J
Processing row 502/1297...
✅ Processed Merge Q363: Answer = C
Processing row 503/1297...
✅ Processed Merge Q363: Answer = B
Processing row 504/1297...
✅ Processed Merge Q363: Answer = F
Processing row 505/1297...
✅ Processed Merge Q363: Answer = B
Processing row 506/1297...
✅ Processed Merge Q363: Answer = I
Processing row 507/1297...
✅ Processed Merge Q363: Answer = G
Processing row 508/1297...
✅ Processed Merge Q363: Answer = D
Processing row 509/1297...
✅ Processed Merge Q363: Answer = I
Processing row 510/1297...
✅ Processed Merge Q363: Answer = J
Saved progress to CSV after 510 items
Processing row 511/1297...
✅ Processed Merge Q363: Answer = B
Processing row 512/1297...
✅ Processed Merge Q363: Answer = H
Processing row 513/1297...
✅ Processed Merge Q363: Answer = J
Processing row 514/1297...
✅ Processed Merge Q363: Answer = D
Processing row 515/1297...
✅ Processed Merge Q363: Answer = C
Processing row 516/1297...
✅ Processed Merge Q363: Answer = D
Processing row 517/1297...
✅ Processed Merge Q363: Answer = D
Processing row 518/1297...
✅ Processed Merge Q363: Answer = I
Processing row 519/1297...
✅ Processed Merge Q363: Answer = G
Processing row 520/1297...
✅ Processed Merge Q363: Answer = A
Saved progress to CSV after 520 items
Processing row 521/1297...
✅ Processed Merge Q363: Answer = C
Processing row 522/1297...
✅ Processed Merge Q363: Answer = C
Processing row 523/1297...
✅ Processed Merge Q363: Answer = I
Processing row 524/1297...
✅ Processed Merge Q363: Answer = A
Processing row 525/1297...
✅ Processed Merge Q363: Answer = B
Processing row 526/1297...
✅ Processed Merge Q363: Answer = C
Processing row 527/1297...
✅ Processed Merge Q363: Answer = D
Processing row 528/1297...
✅ Processed Merge Q363: Answer = D
Processing row 529/1297...
✅ Processed Merge Q363: Answer = F
Processing row 530/1297...
✅ Processed Merge Q363: Answer = H
Saved progress to CSV after 530 items
Processing row 531/1297...
✅ Processed Merge Q363: Answer = A
Processing row 532/1297...
✅ Processed Merge Q363: Answer = C
Processing row 533/1297...
✅ Processed Merge Q363: Answer = D
Processing row 534/1297...
✅ Processed Merge Q363: Answer = B
Processing row 535/1297...
✅ Processed Merge Q363: Answer = C
Processing row 536/1297...
✅ Processed Merge Q363: Answer = C
Processing row 537/1297...
✅ Processed Merge Q363: Answer = D
Processing row 538/1297...
✅ Processed Merge Q363: Answer = B
Processing row 539/1297...
✅ Processed Merge Q363: Answer = G
Processing row 540/1297...
✅ Processed Merge Q363: Answer = B
Saved progress to CSV after 540 items
Processing row 541/1297...
✅ Processed Merge Q363: Answer = A
Processing row 542/1297...
✅ Processed Merge Q363: Answer = H
Processing row 543/1297...
✅ Processed Merge Q363: Answer = B
Processing row 544/1297...
✅ Processed Merge Q363: Answer = H
Processing row 545/1297...
✅ Processed Merge Q363: Answer = D
Processing row 546/1297...
✅ Processed Merge Q363: Answer = D
Processing row 547/1297...
✅ Processed Merge Q363: Answer = E
Processing row 548/1297...
✅ Processed Merge Q363: Answer = D
Processing row 549/1297...
✅ Processed Merge Q363: Answer = F
Processing row 550/1297...
✅ Processed Merge Q363: Answer = I
Saved progress to CSV after 550 items
Processing row 551/1297...
✅ Processed Merge Q363: Answer = C
Processing row 552/1297...
✅ Processed Merge Q363: Answer = B
Processing row 553/1297...
✅ Processed Merge Q363: Answer = C
Processing row 554/1297...
✅ Processed Merge Q363: Answer = E
Processing row 555/1297...
✅ Processed Merge Q363: Answer = D
Processing row 556/1297...
✅ Processed Merge Q363: Answer = C
Processing row 557/1297...
✅ Processed Merge Q363: Answer = C
Processing row 558/1297...
✅ Processed Merge Q363: Answer = B
Processing row 559/1297...
✅ Processed Merge Q363: Answer = D
Processing row 560/1297...
✅ Processed Merge Q363: Answer = A
Saved progress to CSV after 560 items
Processing row 561/1297...
✅ Processed Merge Q363: Answer = D
Processing row 562/1297...
✅ Processed Merge Q363: Answer = D
Processing row 563/1297...
✅ Processed Merge Q363: Answer = C
Processing row 564/1297...
✅ Processed Merge Q363: Answer = I
Processing row 565/1297...
✅ Processed Merge Q363: Answer = C
Processing row 566/1297...
✅ Processed Merge Q363: Answer = D
Processing row 567/1297...
✅ Processed Merge Q363: Answer = A
Processing row 568/1297...
✅ Processed Merge Q363: Answer = C
Processing row 569/1297...
✅ Processed Merge Q363: Answer = C
Processing row 570/1297...
✅ Processed Merge Q363: Answer = G
Saved progress to CSV after 570 items
Processing row 571/1297...
✅ Processed Merge Q363: Answer = C
Processing row 572/1297...
✅ Processed Merge Q363: Answer = D
Processing row 573/1297...
✅ Processed Merge Q363: Answer = C
Processing row 574/1297...
✅ Processed Merge Q363: Answer = G
Processing row 575/1297...
✅ Processed Merge Q363: Answer = B
Processing row 576/1297...
✅ Processed Merge Q363: Answer = B
Processing row 577/1297...
✅ Processed Merge Q363: Answer = D
Processing row 578/1297...
✅ Processed Merge Q363: Answer = C
Processing row 579/1297...
✅ Processed Merge Q363: Answer = B
Processing row 580/1297...
✅ Processed Merge Q363: Answer = F
Saved progress to CSV after 580 items
Processing row 581/1297...
✅ Processed Merge Q363: Answer = F
Processing row 582/1297...
✅ Processed Merge Q363: Answer = F
Processing row 583/1297...
✅ Processed Merge Q363: Answer = D
Processing row 584/1297...
✅ Processed Merge Q363: Answer = C
Processing row 585/1297...
✅ Processed Merge Q363: Answer = B
Processing row 586/1297...
✅ Processed Merge Q363: Answer = C
Processing row 587/1297...
✅ Processed Merge Q363: Answer = B
Processing row 588/1297...
✅ Processed Merge Q363: Answer = G
Processing row 589/1297...
✅ Processed Merge Q363: Answer = F
Processing row 590/1297...
✅ Processed Merge Q363: Answer = I
Saved progress to CSV after 590 items
Processing row 591/1297...
✅ Processed Merge Q363: Answer = B
Processing row 592/1297...
✅ Processed Merge Q363: Answer = I
Processing row 593/1297...
✅ Processed Merge Q363: Answer = I
Processing row 594/1297...
✅ Processed Merge Q363: Answer = H
Processing row 595/1297...
✅ Processed Merge Q363: Answer = A
Processing row 596/1297...
✅ Processed Merge Q363: Answer = A
Processing row 597/1297...
✅ Processed Merge Q363: Answer = B
Processing row 598/1297...
✅ Processed Merge Q363: Answer = A
Processing row 599/1297...
✅ Processed Merge Q363: Answer = C
Processing row 600/1297...
✅ Processed Merge Q363: Answer = E
Saved progress to CSV after 600 items
Processing row 601/1297...
✅ Processed Merge Q363: Answer = E
Processing row 602/1297...
✅ Processed Merge Q363: Answer = A
Processing row 603/1297...
✅ Processed Merge Q363: Answer = A
Processing row 604/1297...
✅ Processed Merge Q363: Answer = D
Processing row 605/1297...
✅ Processed Merge Q363: Answer = B
Processing row 606/1297...
✅ Processed Merge Q363: Answer = H
Processing row 607/1297...
✅ Processed Merge Q363: Answer = D
Processing row 608/1297...
✅ Processed Merge Q363: Answer = H
Processing row 609/1297...
✅ Processed Merge Q363: Answer = B
Processing row 610/1297...
✅ Processed Merge Q363: Answer = B
Saved progress to CSV after 610 items
Processing row 611/1297...
✅ Processed Merge Q363: Answer = C
Processing row 612/1297...
✅ Processed Merge Q363: Answer = B
Processing row 613/1297...
✅ Processed Merge Q363: Answer = F
Processing row 614/1297...
✅ Processed Merge Q363: Answer = D
Processing row 615/1297...
✅ Processed Merge Q363: Answer = D
Processing row 616/1297...
✅ Processed Merge Q363: Answer = D
Processing row 617/1297...
✅ Processed Merge Q363: Answer = C
Processing row 618/1297...
✅ Processed Merge Q363: Answer = C
Processing row 619/1297...
✅ Processed Merge Q363: Answer = F
Processing row 620/1297...
✅ Processed Merge Q363: Answer = B
Saved progress to CSV after 620 items
Processing row 621/1297...
✅ Processed Merge Q363: Answer = D
Processing row 622/1297...
✅ Processed Merge Q363: Answer = D
Processing row 623/1297...
✅ Processed Merge Q363: Answer = J
Processing row 624/1297...
✅ Processed Merge Q363: Answer = A
Processing row 625/1297...
✅ Processed Merge Q363: Answer = I
Processing row 626/1297...
✅ Processed Merge Q363: Answer = D
Processing row 627/1297...
✅ Processed Merge Q363: Answer = D
Processing row 628/1297...
✅ Processed Merge Q363: Answer = H
Processing row 629/1297...
✅ Processed Merge Q363: Answer = B
Processing row 630/1297...
✅ Processed Merge Q363: Answer = A
Saved progress to CSV after 630 items
Processing row 631/1297...
✅ Processed Merge Q363: Answer = B
Processing row 632/1297...
✅ Processed Merge Q363: Answer = J
Processing row 633/1297...
✅ Processed Merge Q363: Answer = C
Processing row 634/1297...
✅ Processed Merge Q363: Answer = J
Processing row 635/1297...
✅ Processed Merge Q363: Answer = C
Processing row 636/1297...
✅ Processed Merge Q363: Answer = B
Processing row 637/1297...
✅ Processed Merge Q363: Answer = C
Processing row 638/1297...
✅ Processed Merge Q363: Answer = C
Processing row 639/1297...
✅ Processed Merge Q363: Answer = B
Processing row 640/1297...
✅ Processed Merge Q363: Answer = B
Saved progress to CSV after 640 items
Processing row 641/1297...
✅ Processed Merge Q363: Answer = J
Processing row 642/1297...
✅ Processed Merge Q363: Answer = B
Processing row 643/1297...
✅ Processed Merge Q363: Answer = I
Processing row 644/1297...
✅ Processed Merge Q363: Answer = F
Processing row 645/1297...
✅ Processed Merge Q363: Answer = G
Processing row 646/1297...
✅ Processed Merge Q363: Answer = F
Processing row 647/1297...
✅ Processed Merge Q363: Answer = B
Processing row 648/1297...
✅ Processed Merge Q363: Answer = F
Processing row 649/1297...
✅ Processed Merge Q363: Answer = E
Processing row 650/1297...
✅ Processed Merge Q363: Answer = I
Saved progress to CSV after 650 items
Processing row 651/1297...
⚠️ Could not extract answer from response for row 651:
Response: To determine the most appropriate next step in managing this patient's postpartum hemorrhage (PPH), ...
✅ Processed Merge Q651: Answer = None
Processing row 652/1297...
✅ Processed Merge Q651: Answer = B
Processing row 653/1297...
✅ Processed Merge Q651: Answer = G
Processing row 654/1297...
✅ Processed Merge Q651: Answer = B
Processing row 655/1297...
✅ Processed Merge Q651: Answer = D
Processing row 656/1297...
✅ Processed Merge Q651: Answer = I
Processing row 657/1297...
✅ Processed Merge Q651: Answer = I
Processing row 658/1297...
✅ Processed Merge Q651: Answer = D
Processing row 659/1297...
✅ Processed Merge Q651: Answer = C
Processing row 660/1297...
✅ Processed Merge Q651: Answer = I
Saved progress to CSV after 660 items
Processing row 661/1297...
✅ Processed Merge Q651: Answer = C
Processing row 662/1297...
✅ Processed Merge Q651: Answer = B
Processing row 663/1297...
✅ Processed Merge Q651: Answer = I
Processing row 664/1297...
✅ Processed Merge Q651: Answer = G
Processing row 665/1297...
✅ Processed Merge Q651: Answer = C
Processing row 666/1297...
✅ Processed Merge Q651: Answer = A
Processing row 667/1297...
✅ Processed Merge Q651: Answer = A
Processing row 668/1297...
✅ Processed Merge Q651: Answer = H
Processing row 669/1297...
✅ Processed Merge Q651: Answer = B
Processing row 670/1297...
✅ Processed Merge Q651: Answer = B
Saved progress to CSV after 670 items
Processing row 671/1297...
✅ Processed Merge Q651: Answer = D
Processing row 672/1297...
✅ Processed Merge Q651: Answer = B
Processing row 673/1297...
✅ Processed Merge Q651: Answer = D
Processing row 674/1297...
✅ Processed Merge Q651: Answer = C
Processing row 675/1297...
✅ Processed Merge Q651: Answer = F
Processing row 676/1297...
✅ Processed Merge Q651: Answer = D
Processing row 677/1297...
✅ Processed Merge Q651: Answer = D
Processing row 678/1297...
✅ Processed Merge Q651: Answer = E
Processing row 679/1297...
✅ Processed Merge Q651: Answer = C
Processing row 680/1297...
✅ Processed Merge Q651: Answer = C
Saved progress to CSV after 680 items
Processing row 681/1297...
✅ Processed Merge Q651: Answer = J
Processing row 682/1297...
✅ Processed Merge Q651: Answer = H
Processing row 683/1297...
✅ Processed Merge Q651: Answer = F
Processing row 684/1297...
✅ Processed Merge Q651: Answer = C
Processing row 685/1297...
✅ Processed Merge Q651: Answer = B
Processing row 686/1297...
✅ Processed Merge Q651: Answer = F
Processing row 687/1297...
✅ Processed Merge Q651: Answer = C
Processing row 688/1297...
✅ Processed Merge Q651: Answer = D
Processing row 689/1297...
✅ Processed Merge Q651: Answer = E
Processing row 690/1297...
✅ Processed Merge Q651: Answer = E
Saved progress to CSV after 690 items
Processing row 691/1297...
✅ Processed Merge Q651: Answer = A
Processing row 692/1297...
✅ Processed Merge Q651: Answer = I
Processing row 693/1297...
⚠️ Could not extract answer from response for row 693:
Response: To determine the most appropriate next step in the patient's management, let's analyze the symptoms ...
✅ Processed Merge Q693: Answer = None
Processing row 694/1297...
✅ Processed Merge Q693: Answer = D
Processing row 695/1297...
✅ Processed Merge Q693: Answer = F
Processing row 696/1297...
⚠️ Could not extract answer from response for row 696:
Response: To determine the most likely diagnosis based on the provided patient case summary, let's analyze the...
✅ Processed Merge Q696: Answer = None
Processing row 697/1297...
✅ Processed Merge Q696: Answer = G
Processing row 698/1297...
✅ Processed Merge Q696: Answer = B
Processing row 699/1297...
✅ Processed Merge Q696: Answer = D
Processing row 700/1297...
✅ Processed Merge Q696: Answer = I
Saved progress to CSV after 700 items
Processing row 701/1297...
✅ Processed Merge Q696: Answer = C
Processing row 702/1297...
✅ Processed Merge Q696: Answer = J
Processing row 703/1297...
✅ Processed Merge Q696: Answer = J
Processing row 704/1297...
✅ Processed Merge Q696: Answer = A
Processing row 705/1297...
✅ Processed Merge Q696: Answer = E
Processing row 706/1297...
✅ Processed Merge Q696: Answer = F
Processing row 707/1297...
✅ Processed Merge Q696: Answer = C
Processing row 708/1297...
✅ Processed Merge Q696: Answer = C
Processing row 709/1297...
✅ Processed Merge Q696: Answer = C
Processing row 710/1297...
✅ Processed Merge Q696: Answer = D
Saved progress to CSV after 710 items
Processing row 711/1297...
✅ Processed Merge Q696: Answer = B
Processing row 712/1297...
✅ Processed Merge Q696: Answer = A
Processing row 713/1297...
✅ Processed Merge Q696: Answer = C
Processing row 714/1297...
✅ Processed Merge Q696: Answer = C
Processing row 715/1297...
✅ Processed Merge Q696: Answer = B
Processing row 716/1297...
✅ Processed Merge Q696: Answer = C
Processing row 717/1297...
✅ Processed Merge Q696: Answer = A
Processing row 718/1297...
✅ Processed Merge Q696: Answer = A
Processing row 719/1297...
✅ Processed Merge Q696: Answer = D
Processing row 720/1297...
✅ Processed Merge Q696: Answer = B
Saved progress to CSV after 720 items
Processing row 721/1297...
✅ Processed Merge Q696: Answer = B
Processing row 722/1297...
✅ Processed Merge Q696: Answer = C
Processing row 723/1297...
✅ Processed Merge Q696: Answer = C
Processing row 724/1297...
✅ Processed Merge Q696: Answer = D
Processing row 725/1297...
✅ Processed Merge Q696: Answer = C
Processing row 726/1297...
✅ Processed Merge Q696: Answer = B
Processing row 727/1297...
✅ Processed Merge Q696: Answer = C
Processing row 728/1297...
✅ Processed Merge Q696: Answer = D
Processing row 729/1297...
✅ Processed Merge Q696: Answer = B
Processing row 730/1297...
✅ Processed Merge Q696: Answer = A
Saved progress to CSV after 730 items
Processing row 731/1297...
✅ Processed Merge Q696: Answer = C
Processing row 732/1297...
✅ Processed Merge Q696: Answer = C
Processing row 733/1297...
✅ Processed Merge Q696: Answer = D
Processing row 734/1297...
✅ Processed Merge Q696: Answer = C
Processing row 735/1297...
✅ Processed Merge Q696: Answer = C
Processing row 736/1297...
✅ Processed Merge Q696: Answer = B
Processing row 737/1297...
✅ Processed Merge Q696: Answer = D
Processing row 738/1297...
✅ Processed Merge Q696: Answer = D
Processing row 739/1297...
✅ Processed Merge Q696: Answer = D
Processing row 740/1297...
✅ Processed Merge Q696: Answer = B
Saved progress to CSV after 740 items
Processing row 741/1297...
✅ Processed Merge Q696: Answer = B
Processing row 742/1297...
✅ Processed Merge Q696: Answer = D
Processing row 743/1297...
✅ Processed Merge Q696: Answer = C
Processing row 744/1297...
✅ Processed Merge Q696: Answer = D
Processing row 745/1297...
✅ Processed Merge Q696: Answer = B
Processing row 746/1297...
✅ Processed Merge Q696: Answer = C
Processing row 747/1297...
✅ Processed Merge Q696: Answer = B
Processing row 748/1297...
✅ Processed Merge Q696: Answer = C
Processing row 749/1297...
✅ Processed Merge Q696: Answer = A
Processing row 750/1297...
✅ Processed Merge Q696: Answer = B
Saved progress to CSV after 750 items
Processing row 751/1297...
✅ Processed Merge Q696: Answer = D
Processing row 752/1297...
✅ Processed Merge Q696: Answer = B
Processing row 753/1297...
✅ Processed Merge Q696: Answer = A
Processing row 754/1297...
✅ Processed Merge Q696: Answer = A
Processing row 755/1297...
✅ Processed Merge Q696: Answer = C
Processing row 756/1297...
✅ Processed Merge Q696: Answer = C
Processing row 757/1297...
✅ Processed Merge Q696: Answer = A
Processing row 758/1297...
✅ Processed Merge Q696: Answer = D
Processing row 759/1297...
✅ Processed Merge Q696: Answer = D
Processing row 760/1297...
✅ Processed Merge Q696: Answer = A
Saved progress to CSV after 760 items
Processing row 761/1297...
✅ Processed Merge Q696: Answer = C
Processing row 762/1297...
✅ Processed Merge Q696: Answer = C
Processing row 763/1297...
✅ Processed Merge Q696: Answer = D
Processing row 764/1297...
✅ Processed Merge Q696: Answer = C
Processing row 765/1297...
✅ Processed Merge Q696: Answer = C
Processing row 766/1297...
✅ Processed Merge Q696: Answer = D
Processing row 767/1297...
✅ Processed Merge Q696: Answer = A
Processing row 768/1297...
✅ Processed Merge Q696: Answer = D
Processing row 769/1297...
✅ Processed Merge Q696: Answer = D
Processing row 770/1297...
✅ Processed Merge Q696: Answer = B
Saved progress to CSV after 770 items
Processing row 771/1297...
✅ Processed Merge Q696: Answer = D
Processing row 772/1297...
✅ Processed Merge Q696: Answer = A
Processing row 773/1297...
✅ Processed Merge Q696: Answer = C
Processing row 774/1297...
✅ Processed Merge Q696: Answer = D
Processing row 775/1297...
✅ Processed Merge Q696: Answer = D
Processing row 776/1297...
✅ Processed Merge Q696: Answer = B
Processing row 777/1297...
✅ Processed Merge Q696: Answer = A
Processing row 778/1297...
✅ Processed Merge Q696: Answer = A
Processing row 779/1297...
✅ Processed Merge Q696: Answer = B
Processing row 780/1297...
✅ Processed Merge Q696: Answer = D
Saved progress to CSV after 780 items
Processing row 781/1297...
✅ Processed Merge Q696: Answer = B
Processing row 782/1297...
✅ Processed Merge Q696: Answer = B
Processing row 783/1297...
✅ Processed Merge Q696: Answer = A
Processing row 784/1297...
✅ Processed Merge Q696: Answer = B
Processing row 785/1297...
✅ Processed Merge Q696: Answer = B
Processing row 786/1297...
✅ Processed Merge Q696: Answer = D
Processing row 787/1297...
✅ Processed Merge Q696: Answer = D
Processing row 788/1297...
✅ Processed Merge Q696: Answer = B
Processing row 789/1297...
✅ Processed Merge Q696: Answer = D
Processing row 790/1297...
✅ Processed Merge Q696: Answer = B
Saved progress to CSV after 790 items
Processing row 791/1297...
✅ Processed Merge Q696: Answer = D
Processing row 792/1297...
✅ Processed Merge Q696: Answer = A
Processing row 793/1297...
✅ Processed Merge Q696: Answer = B
Processing row 794/1297...
✅ Processed Merge Q696: Answer = A
Processing row 795/1297...
✅ Processed Merge Q696: Answer = C
Processing row 796/1297...
✅ Processed Merge Q696: Answer = B
Processing row 797/1297...
✅ Processed Merge Q696: Answer = A
Processing row 798/1297...
✅ Processed Merge Q696: Answer = C
Processing row 799/1297...
✅ Processed Merge Q696: Answer = C
Processing row 800/1297...
✅ Processed Merge Q696: Answer = D
Saved progress to CSV after 800 items
Processing row 801/1297...
✅ Processed Merge Q696: Answer = C
Processing row 802/1297...
✅ Processed Merge Q696: Answer = B
Processing row 803/1297...
✅ Processed Merge Q696: Answer = D
Processing row 804/1297...
✅ Processed Merge Q696: Answer = B
Processing row 805/1297...
✅ Processed Merge Q696: Answer = C
Processing row 806/1297...
✅ Processed Merge Q696: Answer = C
Processing row 807/1297...
✅ Processed Merge Q696: Answer = A
Processing row 808/1297...
✅ Processed Merge Q696: Answer = D
Processing row 809/1297...
✅ Processed Merge Q696: Answer = D
Processing row 810/1297...
✅ Processed Merge Q696: Answer = D
Saved progress to CSV after 810 items
Processing row 811/1297...
✅ Processed Merge Q696: Answer = D
Processing row 812/1297...
✅ Processed Merge Q696: Answer = D
Processing row 813/1297...
✅ Processed Merge Q696: Answer = A
Processing row 814/1297...
✅ Processed Merge Q696: Answer = D
Processing row 815/1297...
✅ Processed Merge Q696: Answer = D
Processing row 816/1297...
✅ Processed Merge Q696: Answer = B
Processing row 817/1297...
✅ Processed Merge Q696: Answer = D
Processing row 818/1297...
✅ Processed Merge Q696: Answer = B
Processing row 819/1297...
✅ Processed Merge Q696: Answer = D
Processing row 820/1297...
✅ Processed Merge Q696: Answer = A
Saved progress to CSV after 820 items
Processing row 821/1297...
✅ Processed Merge Q696: Answer = D
Processing row 822/1297...
✅ Processed Merge Q696: Answer = C
Processing row 823/1297...
✅ Processed Merge Q696: Answer = A
Processing row 824/1297...
✅ Processed Merge Q696: Answer = D
Processing row 825/1297...
✅ Processed Merge Q696: Answer = A
Processing row 826/1297...
✅ Processed Merge Q696: Answer = D
Processing row 827/1297...
✅ Processed Merge Q696: Answer = C
Processing row 828/1297...
✅ Processed Merge Q696: Answer = D
Processing row 829/1297...
✅ Processed Merge Q696: Answer = D
Processing row 830/1297...
✅ Processed Merge Q696: Answer = B
Saved progress to CSV after 830 items
Processing row 831/1297...
✅ Processed Merge Q696: Answer = D
Processing row 832/1297...
✅ Processed Merge Q696: Answer = B
Processing row 833/1297...
✅ Processed Merge Q696: Answer = D
Processing row 834/1297...
✅ Processed Merge Q696: Answer = B
Processing row 835/1297...
✅ Processed Merge Q696: Answer = B
Processing row 836/1297...
✅ Processed Merge Q696: Answer = C
Processing row 837/1297...
✅ Processed Merge Q696: Answer = D
Processing row 838/1297...
✅ Processed Merge Q696: Answer = B
Processing row 839/1297...
✅ Processed Merge Q696: Answer = B
Processing row 840/1297...
✅ Processed Merge Q696: Answer = D
Saved progress to CSV after 840 items
Processing row 841/1297...
✅ Processed Merge Q696: Answer = B
Processing row 842/1297...
✅ Processed Merge Q696: Answer = C
Processing row 843/1297...
✅ Processed Merge Q696: Answer = D
Processing row 844/1297...
✅ Processed Merge Q696: Answer = B
Processing row 845/1297...
✅ Processed Merge Q696: Answer = D
Processing row 846/1297...
✅ Processed Merge Q696: Answer = D
Processing row 847/1297...
✅ Processed Merge Q696: Answer = D
Processing row 848/1297...
✅ Processed Merge Q696: Answer = B
Processing row 849/1297...
✅ Processed Merge Q696: Answer = B
Processing row 850/1297...
✅ Processed Merge Q696: Answer = B
Saved progress to CSV after 850 items
Processing row 851/1297...
✅ Processed Merge Q696: Answer = B
Processing row 852/1297...
✅ Processed Merge Q696: Answer = A
Processing row 853/1297...
✅ Processed Merge Q696: Answer = A
Processing row 854/1297...
✅ Processed Merge Q696: Answer = B
Processing row 855/1297...
✅ Processed Merge Q696: Answer = D
Processing row 856/1297...
✅ Processed Merge Q696: Answer = D
Processing row 857/1297...
✅ Processed Merge Q696: Answer = D
Processing row 858/1297...
✅ Processed Merge Q696: Answer = B
Processing row 859/1297...
✅ Processed Merge Q696: Answer = D
Processing row 860/1297...
✅ Processed Merge Q696: Answer = C
Saved progress to CSV after 860 items
Processing row 861/1297...
✅ Processed Merge Q696: Answer = D
Processing row 862/1297...
✅ Processed Merge Q696: Answer = D
Processing row 863/1297...
✅ Processed Merge Q696: Answer = A
Processing row 864/1297...
✅ Processed Merge Q696: Answer = D
Processing row 865/1297...
✅ Processed Merge Q696: Answer = D
Processing row 866/1297...
✅ Processed Merge Q696: Answer = C
Processing row 867/1297...
✅ Processed Merge Q696: Answer = A
Processing row 868/1297...
✅ Processed Merge Q696: Answer = D
Processing row 869/1297...
✅ Processed Merge Q696: Answer = C
Processing row 870/1297...
✅ Processed Merge Q696: Answer = A
Saved progress to CSV after 870 items
Processing row 871/1297...
✅ Processed Merge Q696: Answer = A
Processing row 872/1297...
✅ Processed Merge Q696: Answer = A
Processing row 873/1297...
✅ Processed Merge Q696: Answer = B
Processing row 874/1297...
✅ Processed Merge Q696: Answer = A
Processing row 875/1297...
✅ Processed Merge Q696: Answer = A
Processing row 876/1297...
✅ Processed Merge Q696: Answer = D
Processing row 877/1297...
✅ Processed Merge Q696: Answer = D
Processing row 878/1297...
✅ Processed Merge Q696: Answer = C
Processing row 879/1297...
✅ Processed Merge Q696: Answer = D
Processing row 880/1297...
✅ Processed Merge Q696: Answer = D
Saved progress to CSV after 880 items
Processing row 881/1297...
✅ Processed Merge Q696: Answer = B
Processing row 882/1297...
✅ Processed Merge Q696: Answer = B
Processing row 883/1297...
✅ Processed Merge Q696: Answer = D
Processing row 884/1297...
✅ Processed Merge Q696: Answer = B
Processing row 885/1297...
✅ Processed Merge Q696: Answer = D
Processing row 886/1297...
✅ Processed Merge Q696: Answer = D
Processing row 887/1297...
✅ Processed Merge Q696: Answer = A
Processing row 888/1297...
✅ Processed Merge Q696: Answer = D
Processing row 889/1297...
✅ Processed Merge Q696: Answer = A
Processing row 890/1297...
✅ Processed Merge Q696: Answer = D
Saved progress to CSV after 890 items
Processing row 891/1297...
✅ Processed Merge Q696: Answer = D
Processing row 892/1297...
✅ Processed Merge Q696: Answer = D
Processing row 893/1297...
✅ Processed Merge Q696: Answer = A
Processing row 894/1297...
✅ Processed Merge Q696: Answer = B
Processing row 895/1297...
✅ Processed Merge Q696: Answer = C
Processing row 896/1297...
✅ Processed Merge Q696: Answer = A
Processing row 897/1297...
✅ Processed Merge Q696: Answer = D
Processing row 898/1297...
✅ Processed Merge Q696: Answer = D
Processing row 899/1297...
✅ Processed Merge Q696: Answer = D
Processing row 900/1297...
✅ Processed Merge Q696: Answer = A
Saved progress to CSV after 900 items
Processing row 901/1297...
✅ Processed Merge Q696: Answer = C
Processing row 902/1297...
✅ Processed Merge Q696: Answer = D
Processing row 903/1297...
✅ Processed Merge Q696: Answer = C
Processing row 904/1297...
✅ Processed Merge Q696: Answer = D
Processing row 905/1297...
✅ Processed Merge Q696: Answer = D
Processing row 906/1297...
✅ Processed Merge Q696: Answer = D
Processing row 907/1297...
✅ Processed Merge Q696: Answer = A
Processing row 908/1297...
✅ Processed Merge Q696: Answer = D
Processing row 909/1297...
✅ Processed Merge Q696: Answer = B
Processing row 910/1297...
✅ Processed Merge Q696: Answer = D
Saved progress to CSV after 910 items
Processing row 911/1297...
✅ Processed Merge Q696: Answer = C
Processing row 912/1297...
✅ Processed Merge Q696: Answer = B
Processing row 913/1297...
✅ Processed Merge Q696: Answer = D
Processing row 914/1297...
✅ Processed Merge Q696: Answer = C
Processing row 915/1297...
✅ Processed Merge Q696: Answer = D
Processing row 916/1297...
✅ Processed Merge Q696: Answer = C
Processing row 917/1297...
✅ Processed Merge Q696: Answer = D
Processing row 918/1297...
✅ Processed Merge Q696: Answer = B
Processing row 919/1297...
✅ Processed Merge Q696: Answer = B
Processing row 920/1297...
✅ Processed Merge Q696: Answer = C
Saved progress to CSV after 920 items
Processing row 921/1297...
✅ Processed Merge Q696: Answer = C
Processing row 922/1297...
✅ Processed Merge Q696: Answer = D
Processing row 923/1297...
✅ Processed Merge Q696: Answer = B
Processing row 924/1297...
✅ Processed Merge Q696: Answer = C
Processing row 925/1297...
✅ Processed Merge Q696: Answer = C
Processing row 926/1297...
✅ Processed Merge Q696: Answer = D
Processing row 927/1297...
✅ Processed Merge Q696: Answer = D
Processing row 928/1297...
✅ Processed Merge Q696: Answer = B
Processing row 929/1297...
✅ Processed Merge Q696: Answer = A
Processing row 930/1297...
✅ Processed Merge Q696: Answer = C
Saved progress to CSV after 930 items
Processing row 931/1297...
✅ Processed Merge Q696: Answer = B
Processing row 932/1297...
✅ Processed Merge Q696: Answer = C
Processing row 933/1297...
✅ Processed Merge Q696: Answer = C
Processing row 934/1297...
✅ Processed Merge Q696: Answer = C
Processing row 935/1297...
✅ Processed Merge Q696: Answer = B
Processing row 936/1297...
✅ Processed Merge Q696: Answer = B
Processing row 937/1297...
✅ Processed Merge Q696: Answer = C
Processing row 938/1297...
✅ Processed Merge Q696: Answer = C
Processing row 939/1297...
✅ Processed Merge Q696: Answer = C
Processing row 940/1297...
✅ Processed Merge Q696: Answer = D
Saved progress to CSV after 940 items
Processing row 941/1297...
✅ Processed Merge Q696: Answer = D
Processing row 942/1297...
✅ Processed Merge Q696: Answer = B
Processing row 943/1297...
✅ Processed Merge Q696: Answer = B
Processing row 944/1297...
✅ Processed Merge Q696: Answer = C
Processing row 945/1297...
✅ Processed Merge Q696: Answer = C
Processing row 946/1297...
✅ Processed Merge Q696: Answer = C
Processing row 947/1297...
✅ Processed Merge Q696: Answer = C
Processing row 948/1297...
✅ Processed Merge Q696: Answer = D
Processing row 949/1297...
✅ Processed Merge Q696: Answer = A
Processing row 950/1297...
✅ Processed Merge Q696: Answer = C
Saved progress to CSV after 950 items
Processing row 951/1297...
✅ Processed Merge Q696: Answer = C
Processing row 952/1297...
✅ Processed Merge Q696: Answer = B
Processing row 953/1297...
✅ Processed Merge Q696: Answer = A
Processing row 954/1297...
✅ Processed Merge Q696: Answer = C
Processing row 955/1297...
✅ Processed Merge Q696: Answer = D
Processing row 956/1297...
✅ Processed Merge Q696: Answer = A
Processing row 957/1297...
✅ Processed Merge Q696: Answer = C
Processing row 958/1297...
✅ Processed Merge Q696: Answer = C
Processing row 959/1297...
✅ Processed Merge Q696: Answer = C
Processing row 960/1297...
✅ Processed Merge Q696: Answer = A
Saved progress to CSV after 960 items
Processing row 961/1297...
✅ Processed Merge Q696: Answer = C
Processing row 962/1297...
✅ Processed Merge Q696: Answer = B
Processing row 963/1297...
✅ Processed Merge Q696: Answer = D
Processing row 964/1297...
✅ Processed Merge Q696: Answer = A
Processing row 965/1297...
✅ Processed Merge Q696: Answer = B
Processing row 966/1297...
✅ Processed Merge Q696: Answer = D
Processing row 967/1297...
✅ Processed Merge Q696: Answer = D
Processing row 968/1297...
✅ Processed Merge Q696: Answer = C
Processing row 969/1297...
✅ Processed Merge Q696: Answer = D
Processing row 970/1297...
✅ Processed Merge Q696: Answer = C
Saved progress to CSV after 970 items
Processing row 971/1297...
✅ Processed Merge Q696: Answer = B
Processing row 972/1297...
✅ Processed Merge Q696: Answer = D
Processing row 973/1297...
✅ Processed Merge Q696: Answer = C
Processing row 974/1297...
✅ Processed Merge Q696: Answer = C
Processing row 975/1297...
✅ Processed Merge Q696: Answer = B
Processing row 976/1297...
✅ Processed Merge Q696: Answer = B
Processing row 977/1297...
✅ Processed Merge Q696: Answer = A
Processing row 978/1297...
✅ Processed Merge Q696: Answer = B
Processing row 979/1297...
✅ Processed Merge Q696: Answer = C
Processing row 980/1297...
✅ Processed Merge Q696: Answer = A
Saved progress to CSV after 980 items
Processing row 981/1297...
✅ Processed Merge Q696: Answer = C
Processing row 982/1297...
✅ Processed Merge Q696: Answer = B
Processing row 983/1297...
✅ Processed Merge Q696: Answer = C
Processing row 984/1297...
✅ Processed Merge Q696: Answer = C
Processing row 985/1297...
✅ Processed Merge Q696: Answer = A
Processing row 986/1297...
✅ Processed Merge Q696: Answer = D
Processing row 987/1297...
✅ Processed Merge Q696: Answer = B
Processing row 988/1297...
✅ Processed Merge Q696: Answer = A
Processing row 989/1297...
✅ Processed Merge Q696: Answer = C
Processing row 990/1297...
✅ Processed Merge Q696: Answer = A
Saved progress to CSV after 990 items
Processing row 991/1297...
✅ Processed Merge Q696: Answer = C
Processing row 992/1297...
✅ Processed Merge Q696: Answer = D
Processing row 993/1297...
✅ Processed Merge Q696: Answer = A
Processing row 994/1297...
✅ Processed Merge Q696: Answer = D
Processing row 995/1297...
✅ Processed Merge Q696: Answer = B
Processing row 996/1297...
✅ Processed Merge Q696: Answer = C
Processing row 997/1297...
✅ Processed Merge Q696: Answer = C
Processing row 998/1297...
✅ Processed Merge Q696: Answer = C
Processing row 999/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1000/1297...
✅ Processed Merge Q696: Answer = C
Saved progress to CSV after 1000 items
Processing row 1001/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1002/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1003/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1004/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1005/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1006/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1007/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1008/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1009/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1010/1297...
✅ Processed Merge Q696: Answer = D
Saved progress to CSV after 1010 items
Processing row 1011/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1012/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1013/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1014/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1015/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1016/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1017/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1018/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1019/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1020/1297...
✅ Processed Merge Q696: Answer = B
Saved progress to CSV after 1020 items
Processing row 1021/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1022/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1023/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1024/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1025/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1026/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1027/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1028/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1029/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1030/1297...
✅ Processed Merge Q696: Answer = C
Saved progress to CSV after 1030 items
Processing row 1031/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1032/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1033/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1034/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1035/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1036/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1037/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1038/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1039/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1040/1297...
✅ Processed Merge Q696: Answer = A
Saved progress to CSV after 1040 items
Processing row 1041/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1042/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1043/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1044/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1045/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1046/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1047/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1048/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1049/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1050/1297...
✅ Processed Merge Q696: Answer = A
Saved progress to CSV after 1050 items
Processing row 1051/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1052/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1053/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1054/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1055/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1056/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1057/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1058/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1059/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1060/1297...
✅ Processed Merge Q696: Answer = C
Saved progress to CSV after 1060 items
Processing row 1061/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1062/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1063/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1064/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1065/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1066/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1067/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1068/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1069/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1070/1297...
✅ Processed Merge Q696: Answer = B
Saved progress to CSV after 1070 items
Processing row 1071/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1072/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1073/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1074/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1075/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1076/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1077/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1078/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1079/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1080/1297...
✅ Processed Merge Q696: Answer = C
Saved progress to CSV after 1080 items
Processing row 1081/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1082/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1083/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1084/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1085/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1086/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1087/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1088/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1089/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1090/1297...
✅ Processed Merge Q696: Answer = B
Saved progress to CSV after 1090 items
Processing row 1091/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1092/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1093/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1094/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1095/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1096/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1097/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1098/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1099/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1100/1297...
✅ Processed Merge Q696: Answer = D
Saved progress to CSV after 1100 items
Processing row 1101/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1102/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1103/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1104/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1105/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1106/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1107/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1108/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1109/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1110/1297...
✅ Processed Merge Q696: Answer = D
Saved progress to CSV after 1110 items
Processing row 1111/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1112/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1113/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1114/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1115/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1116/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1117/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1118/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1119/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1120/1297...
✅ Processed Merge Q696: Answer = B
Saved progress to CSV after 1120 items
Processing row 1121/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1122/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1123/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1124/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1125/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1126/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1127/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1128/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1129/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1130/1297...
✅ Processed Merge Q696: Answer = B
Saved progress to CSV after 1130 items
Processing row 1131/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1132/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1133/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1134/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1135/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1136/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1137/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1138/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1139/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1140/1297...
✅ Processed Merge Q696: Answer = C
Saved progress to CSV after 1140 items
Processing row 1141/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1142/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1143/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1144/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1145/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1146/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1147/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1148/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1149/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1150/1297...
✅ Processed Merge Q696: Answer = A
Saved progress to CSV after 1150 items
Processing row 1151/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1152/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1153/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1154/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1155/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1156/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1157/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1158/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1159/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1160/1297...
✅ Processed Merge Q696: Answer = C
Saved progress to CSV after 1160 items
Processing row 1161/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1162/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1163/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1164/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1165/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1166/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1167/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1168/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1169/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1170/1297...
✅ Processed Merge Q696: Answer = C
Saved progress to CSV after 1170 items
Processing row 1171/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1172/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1173/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1174/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1175/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1176/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1177/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1178/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1179/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1180/1297...
✅ Processed Merge Q696: Answer = B
Saved progress to CSV after 1180 items
Processing row 1181/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1182/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1183/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1184/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1185/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1186/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1187/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1188/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1189/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1190/1297...
✅ Processed Merge Q696: Answer = D
Saved progress to CSV after 1190 items
Processing row 1191/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1192/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1193/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1194/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1195/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1196/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1197/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1198/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1199/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1200/1297...
✅ Processed Merge Q696: Answer = B
Saved progress to CSV after 1200 items
Processing row 1201/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1202/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1203/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1204/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1205/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1206/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1207/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1208/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1209/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1210/1297...
✅ Processed Merge Q696: Answer = A
Saved progress to CSV after 1210 items
Processing row 1211/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1212/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1213/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1214/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1215/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1216/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1217/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1218/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1219/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1220/1297...
✅ Processed Merge Q696: Answer = B
Saved progress to CSV after 1220 items
Processing row 1221/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1222/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1223/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1224/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1225/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1226/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1227/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1228/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1229/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1230/1297...
✅ Processed Merge Q696: Answer = A
Saved progress to CSV after 1230 items
Processing row 1231/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1232/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1233/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1234/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1235/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1236/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1237/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1238/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1239/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1240/1297...
✅ Processed Merge Q696: Answer = C
Saved progress to CSV after 1240 items
Processing row 1241/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1242/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1243/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1244/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1245/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1246/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1247/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1248/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1249/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1250/1297...
✅ Processed Merge Q696: Answer = B
Saved progress to CSV after 1250 items
Processing row 1251/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1252/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1253/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1254/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1255/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1256/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1257/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1258/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1259/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1260/1297...
✅ Processed Merge Q696: Answer = C
Saved progress to CSV after 1260 items
Processing row 1261/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1262/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1263/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1264/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1265/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1266/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1267/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1268/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1269/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1270/1297...
✅ Processed Merge Q696: Answer = B
Saved progress to CSV after 1270 items
Processing row 1271/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1272/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1273/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1274/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1275/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1276/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1277/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1278/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1279/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1280/1297...
✅ Processed Merge Q696: Answer = A
Saved progress to CSV after 1280 items
Processing row 1281/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1282/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1283/1297...
✅ Processed Merge Q696: Answer = A
Processing row 1284/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1285/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1286/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1287/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1288/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1289/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1290/1297...
✅ Processed Merge Q696: Answer = B
Saved progress to CSV after 1290 items
Processing row 1291/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1292/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1293/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1294/1297...
✅ Processed Merge Q696: Answer = C
Processing row 1295/1297...
✅ Processed Merge Q696: Answer = B
Processing row 1296/1297...
✅ Processed Merge Q696: Answer = D
Processing row 1297/1297...
✅ Processed Merge Q696: Answer = D
"""

import pandas as pd
import re

# Load your original data to get the metadata
df_original = pd.read_csv(paths.DATA / "After_Removal_High_qwen_72B_predictions.csv")

print(f"Processing console output...")
print(f"Original dataset has {len(df_original)} rows")

# Extract all the "Processing row X" entries
results = []

# Pattern 1: Successfully extracted answer
# "Processing row X/Y...\n✅ Processed Merge QZ: Answer = LETTER"
pattern_success = r"Processing row (\d+)/\d+\.\.\..*?Answer = ([A-J])"

# Pattern 2: Failed extraction (warning message)
# "Processing row X/Y...\n⚠️ Could not extract answer"
pattern_warning = r"Processing row (\d+)/\d+\.\.\..*?⚠️ Could not extract answer"

# Find all successful extractions
successful_matches = list(re.finditer(pattern_success, console_output, re.DOTALL))
print(f"Found {len(successful_matches)} successful answer extractions")

# Find all failed extractions
failed_matches = list(re.finditer(pattern_warning, console_output, re.DOTALL))
print(f"Found {len(failed_matches)} failed answer extractions")

# Create a dictionary to store all row numbers and their answers
row_answers = {}

# Add successful extractions
for match in successful_matches:
    row_num = int(match.group(1))
    answer = match.group(2).upper()
    row_answers[row_num] = answer

# Add failed extractions (with None as answer)
for match in failed_matches:
    row_num = int(match.group(1))
    # Only add if not already in dictionary (failed extractions come before successful ones in output)
    if row_num not in row_answers:
        row_answers[row_num] = None

print(f"Total unique rows processed: {len(row_answers)}")

# Create results list in order
for row_num in sorted(row_answers.keys()):
    idx = row_num - 1  # Convert to 0-based index
    
    if idx < len(df_original):
        row = df_original.iloc[idx]
        answer = row_answers[row_num]
        
        result_entry = {
            "QA_ID": f"Merge Q{row_num}",
            "Origin": row.get("Origin", ""),
            "data_source": row.get("data_source_corr", ""),
            "Raw_Response": f"(extracted from console - answer: {answer})" if answer else "(no answer extracted)",
            "Extracted_Answer": answer
        }
        results.append(result_entry)
    else:
        print(f"⚠️ Warning: Row {row_num} exceeds original dataset size")
        # Still create entry with empty cell for answer
        result_entry = {
            "QA_ID": f"Merge Q{row_num}",
            "Origin": "",
            "data_source": "",
            "Raw_Response": "(row exceeds original dataset)",
            "Extracted_Answer": ""  # Empty cell
        }
        results.append(result_entry)

# Create DataFrame and save
output_df = pd.DataFrame(results)
output_file = paths.PREDICTIONS / "Qwen_14B_predictions_Qwen72B_FIXED.csv"
output_df.to_csv(output_file, index=False)

print(f"\n{'='*60}")
print(f"✅ Created corrected CSV with {len(results)} rows")
print(f"Saved to: {output_file}")
print(f"\n{'='*60}")
print("Summary:")
print(f"  - Total rows: {len(results)}")
print(f"  - With answers: {output_df['Extracted_Answer'].notna().sum()}")
print(f"  - Without answers (None): {output_df['Extracted_Answer'].isna().sum()}")

print(f"\n{'='*60}")
print("First 10 rows:")
print(output_df.head(10)[['QA_ID', 'Origin', 'Extracted_Answer']])

print(f"\n{'='*60}")
print("Rows with no extracted answer:")
no_answer_rows = output_df[output_df['Extracted_Answer'].isna()]
if len(no_answer_rows) > 0:
    print(no_answer_rows[['QA_ID', 'Origin']])
else:
    print("None - all rows have extracted answers!")
```

```python
import pandas as pd

# Load the data
import pandas as pd

# Load the data
output_df = pd.read_csv(paths.PREDICTIONS / "Qwen_14B_predictions_Qwen72B_FIXED.csv")
df = pd.read_csv(paths.DATA / "After_Removal_High_qwen_72B_predictions.csv")

print("=" * 60)
print("COMPARISON OF 'Origin' COLUMN")
print("=" * 60)

# Basic info
print(f"\nOriginal file rows: {len(df)}")
print(f"Output file rows: {len(output_df)}")
print(f"Missing rows: {len(df) - len(output_df)}")

# Get unique Origin values from each
original_origins = set(df['Origin'].unique())
output_origins = set(output_df['Origin'].unique())

# Find what's missing in output
missing_origins = original_origins - output_origins

print(f"\n" + "=" * 60)
print("UNIQUE 'Origin' VALUES")
print("=" * 60)
print(f"\nUnique Origins in original file: {len(original_origins)}")
print(f"Unique Origins in output file: {len(output_origins)}")

if missing_origins:
    print(f"\n⚠️  MISSING Origin values in output file: {len(missing_origins)}")
    print("\nMissing Origin values:")
    for origin in sorted(missing_origins):
        count = len(df[df['Origin'] == origin])
        print(f"  - '{origin}' (appears {count} times in original)")
else:
    print("\n✅ No missing Origin values - all unique Origins are present!")

# Check for extra origins in output (shouldn't happen, but good to check)
extra_origins = output_origins - original_origins
if extra_origins:
    print(f"\n⚠️  EXTRA Origin values in output file (shouldn't exist): {len(extra_origins)}")
    for origin in sorted(extra_origins):
        print(f"  - '{origin}'")

# Detailed row-by-row comparison
print(f"\n" + "=" * 60)
print("ROW-BY-ROW MISSING DATA")
print("=" * 60)

if len(output_df) < len(df):
    missing_rows = df.iloc[len(output_df):]
    print(f"\nRows {len(output_df) + 1} to {len(df)} are missing from output:")
    print("\nOrigin values of missing rows:")
    print(missing_rows['Origin'].value_counts())
    
    print("\nFirst 10 missing rows:")
    print(missing_rows[['Origin']].head(10))

# Summary statistics
print(f"\n" + "=" * 60)
print("ORIGIN VALUE COUNTS COMPARISON")
print("=" * 60)

print("\nOriginal file Origin counts:")
print(df['Origin'].value_counts().sort_index())

print("\nOutput file Origin counts:")
print(output_df['Origin'].value_counts().sort_index())

# Calculate difference
print("\nDifference (Original - Output):")
original_counts = df['Origin'].value_counts()
output_counts = output_df['Origin'].value_counts()

all_origins = set(original_counts.index) | set(output_counts.index)
for origin in sorted(all_origins):
    orig_count = original_counts.get(origin, 0)
    out_count = output_counts.get(origin, 0)
    diff = orig_count - out_count
    if diff != 0:
        print(f"  {origin}: {diff} rows missing")
    
```

```python
import pandas as pd
import numpy as np
from scipy import stats

# Load the data
output_df = pd.read_csv(paths.PREDICTIONS / "Qwen_14B_predictions_Qwen72B_FIXED.csv")
df = pd.read_csv(paths.DATA / "After_Removal_High_qwen_72B_predictions.csv")

# Ensure both dataframes have the same length
assert len(output_df) == len(df), "DataFrames have different lengths!"

# Calculate accuracy (assuming both columns contain the same type of answers to compare)
# Method 1: Exact match
output_df['match'] = (output_df['Extracted_Answer'] == df['answer_df3']).astype(int)

# Overall accuracy statistics
accuracy = output_df['match'].mean()
std_dev = output_df['match'].std()
n = len(output_df)
se = std_dev / np.sqrt(n)  # Standard error
ci_95 = stats.t.interval(0.95, n-1, loc=accuracy, scale=se)

print("=" * 60)
print("OVERALL ACCURACY ANALYSIS")
print("=" * 60)
print(f"Accuracy: {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"Standard Deviation: {std_dev:.4f}")
print(f"95% Confidence Interval: [{ci_95[0]:.4f}, {ci_95[1]:.4f}]")
print(f"95% CI (percentage): [{ci_95[0]*100:.2f}%, {ci_95[1]*100:.2f}%]")
print(f"Sample Size: {n}")
print("=" * 60)
print()

# Analysis by category (assuming data_source_corr is in one of the dataframes)
# Check which dataframe has data_source_corr
if 'data_source_corr' in output_df.columns:
    analysis_df = output_df.copy()
elif 'data_source_corr' in df.columns:
    analysis_df = output_df.copy()
    analysis_df['data_source_corr'] = df['data_source_corr']
else:
    print("Warning: 'data_source_corr' column not found in either dataframe")
    analysis_df = output_df.copy()

# Category-wise analysis
if 'data_source_corr' in analysis_df.columns:
    category_stats = analysis_df.groupby('data_source_corr')['match'].agg([
        ('Count', 'count'),
        ('Mean_Accuracy', 'mean'),
        ('Std_Dev', 'std'),
        ('SE', lambda x: x.std() / np.sqrt(len(x)))
    ]).reset_index()
    
    # Calculate 95% CI for each category
    ci_lower = []
    ci_upper = []
    
    for idx, row in category_stats.iterrows():
        n_cat = row['Count']
        mean_cat = row['Mean_Accuracy']
        se_cat = row['SE']
        
        if n_cat > 1:
            ci = stats.t.interval(0.95, n_cat-1, loc=mean_cat, scale=se_cat)
            ci_lower.append(ci[0])
            ci_upper.append(ci[1])
        else:
            ci_lower.append(np.nan)
            ci_upper.append(np.nan)
    
    category_stats['CI_95_Lower'] = ci_lower
    category_stats['CI_95_Upper'] = ci_upper
    
    # Format percentages
    category_stats['Mean_Accuracy_%'] = category_stats['Mean_Accuracy'] * 100
    category_stats['Std_Dev_%'] = category_stats['Std_Dev'] * 100
    category_stats['CI_95_Lower_%'] = category_stats['CI_95_Lower'] * 100
    category_stats['CI_95_Upper_%'] = category_stats['CI_95_Upper'] * 100
    
    print("CATEGORY-WISE ACCURACY ANALYSIS")
    print("=" * 60)
    print(category_stats.to_string(index=False))
    print("=" * 60)
    print()
    
# Create summary statistics table
summary_table = pd.DataFrame({
    'Metric': ['Overall Accuracy', 'Standard Deviation', '95% CI Lower', '95% CI Upper', 'Sample Size'],
    'Value': [f"{accuracy:.4f} ({accuracy*100:.2f}%)", 
              f"{std_dev:.4f}", 
              f"{ci_95[0]:.4f} ({ci_95[0]*100:.2f}%)", 
              f"{ci_95[1]:.4f} ({ci_95[1]*100:.2f}%)", 
              n]
})

print("\nSUMMARY TABLE")
print("=" * 60)
print(summary_table.to_string(index=False))
print("=" * 60)
```
